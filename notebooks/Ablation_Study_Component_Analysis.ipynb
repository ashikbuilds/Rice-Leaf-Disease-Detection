```python
# ================================================================
# RLDAF-Net | FAST DUAL-CNN FEATURE-CACHED PIPELINE
# Rice Leaf Disease Classification | Kaggle | 2 x NVIDIA T4
#
# Backbone 1 : ConvNeXtV2 Tiny
# Backbone 2 : EfficientNetV2-S
#
# IMPORTANT DESIGN:
# 1. Split FIRST, augmentation ONLY on training data.
# 2. Both CNN backbones are pretrained + frozen.
# 3. CNN features are extracted ONCE and cached.
# 4. Training epochs operate only on cached features.
# 5. Therefore the expensive CNN forward pass is NOT repeated
#    for all 10 epochs.
#
# Outputs:
# Accuracy/Loss curves, Macro-F1, ROC-AUC, PR curves,
# Confusion matrices, Classification report, Bootstrap CI,
# McNemar test, calibration/ECE, confidence analysis,
# per-class analysis, t-SNE, robustness analysis,
# model complexity, predictions and saved model.
# ================================================================

import os, sys, gc, json, time, random, warnings, copy, math
warnings.filterwarnings("ignore")

# -------------------- PACKAGE CHECK --------------------
def ensure_package(import_name, package_name):
    try:
        __import__(import_name)
    except ImportError:
        os.system(f"{sys.executable} -m pip install -q {package_name}")

ensure_package("timm", "timm")
ensure_package("albumentations", "albumentations")
ensure_package("sklearn", "scikit-learn")

# -------------------- IMPORTS --------------------
import numpy as np
import pandas as pd
from PIL import Image, ImageFile

ImageFile.LOAD_TRUNCATED_IMAGES = True

import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, TensorDataset

import timm
import albumentations as A
from albumentations.pytorch import ToTensorV2

from tqdm.auto import tqdm

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    matthews_corrcoef, confusion_matrix, classification_report,
    roc_auc_score, roc_curve, precision_recall_curve,
    average_precision_score
)
from sklearn.preprocessing import label_binarize
from sklearn.manifold import TSNE

# ================================================================
# 1. CONFIGURATION
# ================================================================

SEED = 42

CFG = {
    "image_size": 224,

    # Feature extraction batch can be larger because no gradients
    # are stored. Automatically reduced if CUDA OOM occurs.
    "feature_batch_size": 32,

    # Fusion head is very light, so use a large batch.
    "fusion_batch_size": 256,

    "epochs": 10,
    "lr": 3e-4,
    "weight_decay": 1e-4,

    "projection_dim": 512,
    "hidden_dim": 512,
    "dropout": 0.30,

    "label_smoothing": 0.05,

    # Number of augmented views for each TRAIN image.
    # 2 gives stronger training while keeping cache small.
    "train_views": 2,

    "num_workers": 2,
    "pin_memory": True,

    "use_amp": True,

    "bootstrap_iterations": 1000,

    "output_dir": "/kaggle/working/RLDAF_DualCNN_FAST_RESULTS",
}

os.makedirs(CFG["output_dir"], exist_ok=True)

def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)

    # Faster convolution kernels on fixed image size.
    torch.backends.cudnn.benchmark = True
    torch.backends.cudnn.deterministic = False

seed_everything(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
GPU_COUNT = torch.cuda.device_count() if torch.cuda.is_available() else 0

print("=" * 95)
print("RLDAF-Net | FAST DUAL-CNN FEATURE-CACHED PIPELINE")
print("=" * 95)
print("PyTorch:", torch.__version__)
print("Device:", device)
print("GPU count:", GPU_COUNT)

for i in range(GPU_COUNT):
    print(f"GPU {i}: {torch.cuda.get_device_name(i)}")

if GPU_COUNT:
    print("GPU memory:")
    for i in range(GPU_COUNT):
        props = torch.cuda.get_device_properties(i)
        print(f"  GPU {i}: {props.total_memory / 1024**3:.2f} GB")

print("=" * 95)

# ================================================================
# 2. DATASET DISCOVERY
# ================================================================

candidate_roots = [
    "/kaggle/input/datasets/pranzalpranmoni/riceleaf-disease-detection/RiceLeafDiseaseBD/Original images",
    "/kaggle/input/riceleaf-disease-detection/RiceLeafDiseaseBD/Original images",
    "/kaggle/input/RiceLeafDiseaseBD/Original images",
    "/kaggle/input/riceleafdiseasebd/RiceLeafDiseaseBD/Original images",
]

DATA_ROOT = None

for p in candidate_roots:
    if os.path.isdir(p):
        DATA_ROOT = p
        break

if DATA_ROOT is None:
    for root, dirs, files in os.walk("/kaggle/input"):
        if os.path.basename(root).strip().lower() == "original images":
            DATA_ROOT = root
            break

if DATA_ROOT is None:
    raise FileNotFoundError(
        "Could not find 'Original images' inside /kaggle/input. "
        "Please check the Kaggle dataset path."
    )

VALID_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

class_names = sorted([
    d for d in os.listdir(DATA_ROOT)
    if os.path.isdir(os.path.join(DATA_ROOT, d))
])

if len(class_names) < 2:
    raise RuntimeError("Less than two disease classes were found.")

class_to_idx = {name: i for i, name in enumerate(class_names)}
idx_to_class = {i: name for name, i in class_to_idx.items()}
NUM_CLASSES = len(class_names)

records = []

for class_name in class_names:
    class_dir = os.path.join(DATA_ROOT, class_name)

    for root, _, files in os.walk(class_dir):
        for file_name in files:
            if file_name.lower().endswith(VALID_EXTENSIONS):
                records.append({
                    "path": os.path.join(root, file_name),
                    "filename": file_name,
                    "class_name": class_name,
                    "label": class_to_idx[class_name]
                })

df = pd.DataFrame(records)

if df.empty:
    raise RuntimeError("No image files were discovered.")

print("\nDataset root:")
print(DATA_ROOT)

print("\nClasses:")
for i, c in enumerate(class_names):
    print(f"{i}: {c}")

print("\nTotal images found:", len(df))

print("\nClass distribution:")
print(df["class_name"].value_counts().sort_index())

# ================================================================
# 3. IMAGE VALIDATION
# ================================================================

def check_image(path):
    try:
        with Image.open(path) as img:
            img.verify()
        return True
    except Exception:
        return False

print("\nValidating images...")

valid_flags = []
for p in tqdm(df["path"].tolist(), desc="Image validation"):
    valid_flags.append(check_image(p))

df["valid"] = valid_flags

invalid_count = int((~df["valid"]).sum())
df = df[df["valid"]].drop(columns=["valid"]).reset_index(drop=True)

print("Invalid images:", invalid_count)
print("Valid images:", len(df))

if len(df) < 100:
    raise RuntimeError("Too few valid images after validation.")

# ================================================================
# 4. STRATIFIED DATA SPLIT
# IMPORTANT: SPLIT BEFORE AUGMENTATION
# ================================================================

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["label"],
    random_state=SEED
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=SEED
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("\n" + "=" * 95)
print("DATA SPLIT")
print("=" * 95)
print("Train      :", len(train_df))
print("Validation :", len(val_df))
print("Test       :", len(test_df))

print("\nTrain distribution:")
print(train_df["class_name"].value_counts().sort_index())

print("\nValidation distribution:")
print(val_df["class_name"].value_counts().sort_index())

print("\nTest distribution:")
print(test_df["class_name"].value_counts().sort_index())

# Save split metadata.
train_df.to_csv(os.path.join(CFG["output_dir"], "train_split.csv"), index=False)
val_df.to_csv(os.path.join(CFG["output_dir"], "validation_split.csv"), index=False)
test_df.to_csv(os.path.join(CFG["output_dir"], "test_split.csv"), index=False)

# ================================================================
# 5. PREPROCESSING / AUGMENTATION
# ================================================================

MEAN = (0.485, 0.456, 0.406)
STD = (0.229, 0.224, 0.225)

# Training augmentation.
# Applied ONLY to training images.
train_transform = A.Compose([
    A.RandomResizedCrop(
        size=(CFG["image_size"], CFG["image_size"]),
        scale=(0.80, 1.0),
        ratio=(0.90, 1.10),
        p=1.0
    ),
    A.HorizontalFlip(p=0.50),
    A.VerticalFlip(p=0.10),

    A.Affine(
        scale=(0.90, 1.10),
        translate_percent=(-0.05, 0.05),
        rotate=(-15, 15),
        shear=(-5, 5),
        p=0.25
    ),

    A.OneOf([
        A.RandomBrightnessContrast(
            brightness_limit=0.15,
            contrast_limit=0.15
        ),
        A.CLAHE(clip_limit=2.0),
        A.HueSaturationValue(
            hue_shift_limit=6,
            sat_shift_limit=10,
            val_shift_limit=10
        )
    ], p=0.35),

    A.OneOf([
        A.GaussianBlur(blur_limit=(3, 5)),
        A.MotionBlur(blur_limit=5),
        A.GaussNoise(std_range=(0.01, 0.04))
    ], p=0.10),

    A.Normalize(mean=MEAN, std=STD),
    ToTensorV2()
])

# Deterministic evaluation preprocessing.
eval_transform = A.Compose([
    A.Resize(
        height=CFG["image_size"],
        width=CFG["image_size"]
    ),
    A.Normalize(mean=MEAN, std=STD),
    ToTensorV2()
])

# ================================================================
# 6. DATASET CLASS
# ================================================================

class RiceImageDataset(Dataset):

    def __init__(self, dataframe, transform):
        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):
        row = self.df.iloc[index]

        try:
            image = np.array(
                Image.open(row["path"]).convert("RGB")
            )
        except Exception:
            # Extremely defensive fallback.
            image = np.zeros(
                (CFG["image_size"], CFG["image_size"], 3),
                dtype=np.uint8
            )

        image = self.transform(image=image)["image"]

        return image, int(row["label"])


# ================================================================
# 7. LOAD TWO CNN BACKBONES
# ================================================================

print("\n" + "=" * 95)
print("LOADING PRETRAINED CNN BACKBONES")
print("=" * 95)

def create_backbone(model_name):
    try:
        print(f"\nLoading {model_name} with pretrained weights...")
        model = timm.create_model(
            model_name,
            pretrained=True,
            num_classes=0,
            global_pool="avg"
        )
        print("Pretrained weights loaded.")
        return model

    except Exception as e:
        print("\nWARNING: pretrained weights could not be loaded.")
        print("Reason:", str(e)[:300])
        print("Falling back to pretrained=False.")
        print("For the intended research experiment, Kaggle Internet")
        print("should be enabled so pretrained weights can download.")

        model = timm.create_model(
            model_name,
            pretrained=False,
            num_classes=0,
            global_pool="avg"
        )
        return model


cnn1 = create_backbone(
    "convnextv2_tiny.fcmae_ft_in22k_in1k"
)

cnn2 = create_backbone(
    "tf_efficientnetv2_s.in21k_ft_in1k"
)

cnn1.eval()
cnn2.eval()

for p in cnn1.parameters():
    p.requires_grad = False

for p in cnn2.parameters():
    p.requires_grad = False

cnn1 = cnn1.to(device)
cnn2 = cnn2.to(device)

# Do NOT use DataParallel for the fusion head.
# For feature extraction, DataParallel is optional. The code below
# uses a single GPU per process for maximum stability and simplicity.
#
# The major speed improvement comes from caching the CNN features.
# CNNs therefore run only once/twice during feature extraction,
# not during all 10 training epochs.

CNN1_DIM = cnn1.num_features
CNN2_DIM = cnn2.num_features

print("\nBackbone feature dimensions:")
print("ConvNeXtV2:", CNN1_DIM)
print("EfficientNetV2:", CNN2_DIM)

# ================================================================
# 8. FEATURE EXTRACTION
# ================================================================

def make_loader(dataframe, transform, batch_size, shuffle=False):
    ds = RiceImageDataset(dataframe, transform)

    kwargs = {
        "batch_size": batch_size,
        "shuffle": shuffle,
        "num_workers": CFG["num_workers"],
        "pin_memory": CFG["pin_memory"] and torch.cuda.is_available()
    }

    if CFG["num_workers"] > 0:
        kwargs["persistent_workers"] = True
        kwargs["prefetch_factor"] = 2

    return DataLoader(ds, **kwargs)


@torch.inference_mode()
def extract_features(dataframe, transform, views=1, split_name="data"):
    """
    Extract features from both frozen CNNs.

    Returns:
        combined features [N, CNN1_DIM + CNN2_DIM]
        labels [N]
    """

    batch_size = CFG["feature_batch_size"]

    # If GPU memory is insufficient, automatically reduce the batch.
    while True:
        try:
            loader = make_loader(
                dataframe,
                transform,
                batch_size,
                shuffle=False
            )

            all_features = []
            all_labels = []

            print(
                f"\nExtracting {split_name} features | "
                f"views={views} | batch={batch_size}"
            )

            for view in range(views):

                if views > 1:
                    print(f"\nAugmented feature view {view + 1}/{views}")

                view_features = []
                view_labels = []

                for images, labels in tqdm(
                    loader,
                    desc=f"{split_name} view {view+1}",
                    leave=False
                ):
                    images = images.to(
                        device,
                        non_blocking=True
                    )

                    with torch.autocast(
                        device_type="cuda",
                        dtype=torch.float16,
                        enabled=CFG["use_amp"] and torch.cuda.is_available()
                    ):
                        f1 = cnn1(images)
                        f2 = cnn2(images)

                    f1 = f1.float().cpu()
                    f2 = f2.float().cpu()

                    combined = torch.cat([f1, f2], dim=1)

                    view_features.append(combined)
                    view_labels.append(labels.clone())

                    del images, f1, f2, combined

                view_features = torch.cat(view_features, dim=0)
                view_labels = torch.cat(view_labels, dim=0)

                all_features.append(view_features)
                all_labels.append(view_labels)

                del view_features, view_labels
                gc.collect()

                if torch.cuda.is_available():
                    torch.cuda.empty_cache()

            features = torch.cat(all_features, dim=0)
            labels = torch.cat(all_labels, dim=0)

            del all_features, all_labels
            gc.collect()

            return features, labels

        except torch.cuda.OutOfMemoryError:

            if torch.cuda.is_available():
                torch.cuda.empty_cache()

            gc.collect()

            if batch_size <= 4:
                raise RuntimeError(
                    "CUDA OOM even with feature extraction batch <= 4."
                )

            batch_size = max(batch_size // 2, 4)

            print(
                f"\nCUDA OOM detected. Retrying feature extraction "
                f"with batch size {batch_size}..."
            )


# ================================================================
# 9. EXTRACT + CACHE FEATURES
# ================================================================

cache_path = os.path.join(
    CFG["output_dir"],
    "feature_cache.pt"
)

if os.path.exists(cache_path):
    print("\nExisting feature cache found.")
    print("Loading cached features...")
    cache = torch.load(
        cache_path,
        map_location="cpu",
        weights_only=False
    )

    X_train = cache["X_train"]
    y_train = cache["y_train"]
    X_val = cache["X_val"]
    y_val = cache["y_val"]
    X_test = cache["X_test"]
    y_test = cache["y_test"]

else:
    print("\n" + "=" * 95)
    print("FEATURE EXTRACTION AND CACHING")
    print("=" * 95)

    extraction_start = time.time()

    # Two stochastic views for TRAIN.
    X_train, y_train = extract_features(
        train_df,
        train_transform,
        views=CFG["train_views"],
        split_name="TRAIN"
    )

    # Validation/test remain deterministic.
    X_val, y_val = extract_features(
        val_df,
        eval_transform,
        views=1,
        split_name="VALIDATION"
    )

    X_test, y_test = extract_features(
        test_df,
        eval_transform,
        views=1,
        split_name="TEST"
    )

    cache = {
        "X_train": X_train.half(),
        "y_train": y_train.long(),
        "X_val": X_val.half(),
        "y_val": y_val.long(),
        "X_test": X_test.half(),
        "y_test": y_test.long(),
        "class_names": class_names,
        "cnn1_dim": CNN1_DIM,
        "cnn2_dim": CNN2_DIM
    }

    torch.save(cache, cache_path)

    print(
        "\nFeature extraction completed in "
        f"{(time.time() - extraction_start)/60:.2f} minutes."
    )

# Delete frozen CNNs after feature caching.
# This frees GPU memory before fusion-head training.
del cnn1
del cnn2
gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()

print("\nFeature shapes:")
print("Train:", tuple(X_train.shape))
print("Val  :", tuple(X_val.shape))
print("Test :", tuple(X_test.shape))

# ================================================================
# 10. FEATURE NORMALIZATION
# ================================================================
# Standardize each feature dimension using TRAIN statistics only.
# This avoids information leakage.

feature_mean = X_train.float().mean(dim=0)
feature_std = X_train.float().std(dim=0).clamp_min(1e-6)

X_train = ((X_train.float() - feature_mean) / feature_std).half()
X_val = ((X_val.float() - feature_mean) / feature_std).half()
X_test = ((X_test.float() - feature_mean) / feature_std).half()

torch.save(
    {
        "mean": feature_mean.half(),
        "std": feature_std.half()
    },
    os.path.join(CFG["output_dir"], "feature_normalization.pt")
)

# ================================================================
# 11. CLASS WEIGHTS
# ================================================================

class_weights_np = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(NUM_CLASSES),
    y=y_train.numpy()
)

class_weights = torch.tensor(
    class_weights_np,
    dtype=torch.float32,
    device=device
)

print("\nClass weights:")
for i, c in enumerate(class_names):
    print(f"{c:20s}: {class_weights_np[i]:.4f}")

# ================================================================
# 12. FAST DUAL-CNN FUSION NETWORK
# ================================================================

class GatedFusionBlock(nn.Module):

    def __init__(self, dim, dropout=0.30):
        super().__init__()

        self.gate = nn.Sequential(
            nn.Linear(dim * 2, dim),
            nn.LayerNorm(dim),
            nn.GELU(),
            nn.Linear(dim, dim),
            nn.Sigmoid()
        )

        self.mix = nn.Sequential(
            nn.Linear(dim * 2, dim),
            nn.LayerNorm(dim),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(dim, dim)
        )

    def forward(self, a, b):

        cat = torch.cat([a, b], dim=1)

        gate = self.gate(cat)

        mixed = self.mix(cat)

        fused = gate * a + (1.0 - gate) * b

        fused = fused + mixed

        return fused, gate


class DiseaseAwareFusionNet(nn.Module):

    def __init__(
        self,
        input_dim,
        num_classes,
        projection_dim=512,
        hidden_dim=512,
        dropout=0.30
    ):
        super().__init__()

        # Split concatenated feature into two CNN features.
        self.cnn1_dim = CNN1_DIM
        self.cnn2_dim = CNN2_DIM

        self.proj1 = nn.Sequential(
            nn.Linear(self.cnn1_dim, projection_dim),
            nn.LayerNorm(projection_dim),
            nn.GELU(),
            nn.Dropout(dropout)
        )

        self.proj2 = nn.Sequential(
            nn.Linear(self.cnn2_dim, projection_dim),
            nn.LayerNorm(projection_dim),
            nn.GELU(),
            nn.Dropout(dropout)
        )

        # Three-scale feature construction.
        self.scale1 = nn.Sequential(
            nn.Linear(projection_dim, projection_dim),
            nn.LayerNorm(projection_dim),
            nn.GELU()
        )

        self.scale2 = nn.Sequential(
            nn.Linear(projection_dim, projection_dim),
            nn.LayerNorm(projection_dim),
            nn.GELU()
        )

        self.scale3 = nn.Sequential(
            nn.Linear(projection_dim, projection_dim),
            nn.LayerNorm(projection_dim),
            nn.GELU()
        )

        self.fusion = GatedFusionBlock(
            projection_dim,
            dropout
        )

        # Disease-aware channel attention.
        reduction = max(projection_dim // 4, 32)

        self.disease_attention = nn.Sequential(
            nn.LayerNorm(projection_dim),
            nn.Linear(projection_dim, reduction),
            nn.GELU(),
            nn.Linear(reduction, projection_dim),
            nn.Sigmoid()
        )

        self.feature_norm = nn.LayerNorm(projection_dim)

        self.classifier = nn.Sequential(
            nn.Linear(projection_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 256),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(256, num_classes)
        )

        self.embedding_head = nn.Sequential(
            nn.Linear(projection_dim, 256),
            nn.GELU(),
            nn.Linear(256, 128)
        )

    def forward(self, combined, return_features=False):

        a = combined[:, :self.cnn1_dim]
        b = combined[:, self.cnn1_dim:]

        a = self.proj1(a)
        b = self.proj2(b)

        # Multi-scale representation.
        s1 = self.scale1(a)

        s2 = self.scale2(
            (a + b) / 2.0
        )

        s3 = self.scale3(b)

        multi_scale = (
            s1 + s2 + s3
        ) / 3.0

        fused, gate = self.fusion(
            a,
            b
        )

        fused = fused + multi_scale

        disease_weights = self.disease_attention(
            fused
        )

        fused = fused * disease_weights

        fused = self.feature_norm(fused)

        logits = self.classifier(fused)

        embedding = F.normalize(
            self.embedding_head(fused),
            dim=1
        )

        if return_features:
            return {
                "logits": logits,
                "features": fused,
                "embedding": embedding,
                "fusion_gate": gate,
                "disease_attention": disease_weights
            }

        return logits


model = DiseaseAwareFusionNet(
    input_dim=X_train.shape[1],
    num_classes=NUM_CLASSES,
    projection_dim=CFG["projection_dim"],
    hidden_dim=CFG["hidden_dim"],
    dropout=CFG["dropout"]
).to(device)

# ================================================================
# 13. MODEL COMPLEXITY
# ================================================================

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print("\n" + "=" * 95)
print("FUSION MODEL")
print("=" * 95)
print("Total fusion parameters    :", f"{total_params/1e6:.3f} M")
print("Trainable fusion parameters:", f"{trainable_params/1e6:.3f} M")
print("CNN features:", X_train.shape[1])
print("=" * 95)

# ================================================================
# 14. DATA LOADERS FOR FAST FUSION TRAINING
# ================================================================

train_tensor_ds = TensorDataset(
    X_train,
    y_train
)

val_tensor_ds = TensorDataset(
    X_val,
    y_val
)

test_tensor_ds = TensorDataset(
    X_test,
    y_test
)

train_loader = DataLoader(
    train_tensor_ds,
    batch_size=CFG["fusion_batch_size"],
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

val_loader = DataLoader(
    val_tensor_ds,
    batch_size=CFG["fusion_batch_size"],
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

test_loader = DataLoader(
    test_tensor_ds,
    batch_size=CFG["fusion_batch_size"],
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

print("\nFusion training batches:")
print("Train:", len(train_loader))
print("Val  :", len(val_loader))
print("Test :", len(test_loader))

# ================================================================
# 15. LOSS / OPTIMIZER
# ================================================================

criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    label_smoothing=CFG["label_smoothing"]
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=CFG["lr"],
    weight_decay=CFG["weight_decay"]
)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=CFG["epochs"]
)

try:
    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=CFG["use_amp"] and torch.cuda.is_available()
    )
except Exception:
    scaler = torch.cuda.amp.GradScaler(
        enabled=CFG["use_amp"] and torch.cuda.is_available()
    )

# ================================================================
# 16. METRIC FUNCTIONS
# ================================================================

def calculate_metrics(y_true, y_pred):

    return {
        "accuracy": accuracy_score(
            y_true, y_pred
        ),

        "precision": precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "recall": recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "macro_f1": f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "weighted_f1": f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        ),

        "mcc": matthews_corrcoef(
            y_true,
            y_pred
        )
    }


# ================================================================
# 17. TRAIN ONE EPOCH
# ================================================================

def train_one_epoch():

    model.train()

    total_loss = 0.0
    total_correct = 0
    total_samples = 0

    all_true = []
    all_pred = []

    amp_enabled = (
        CFG["use_amp"]
        and torch.cuda.is_available()
    )

    progress = tqdm(
        train_loader,
        desc="Training",
        leave=False
    )

    for features, labels in progress:

        features = features.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=amp_enabled
        ):

            logits = model(features)

            loss = criterion(
                logits,
                labels
            )

        if amp_enabled:

            scaler.scale(loss).backward()

            scaler.unscale_(optimizer)

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                max_norm=1.0
            )

            scaler.step(optimizer)
            scaler.update()

        else:

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                max_norm=1.0
            )

            optimizer.step()

        predictions = logits.argmax(dim=1)

        batch_size = labels.size(0)

        total_loss += (
            loss.item() * batch_size
        )

        total_correct += (
            predictions == labels
        ).sum().item()

        total_samples += batch_size

        all_true.extend(
            labels.detach().cpu().numpy()
        )

        all_pred.extend(
            predictions.detach().cpu().numpy()
        )

        progress.set_postfix(
            loss=f"{loss.item():.4f}",
            acc=f"{total_correct/max(total_samples,1):.4f}"
        )

    y = np.asarray(all_true)
    p = np.asarray(all_pred)

    metrics = calculate_metrics(y, p)

    metrics["loss"] = (
        total_loss / max(total_samples, 1)
    )

    return metrics


# ================================================================
# 18. VALIDATION / TEST
# ================================================================

@torch.inference_mode()
def evaluate(loader, description="Evaluation"):

    model.eval()

    total_loss = 0.0
    total_samples = 0

    all_true = []
    all_pred = []
    all_prob = []
    all_features = []

    amp_enabled = (
        CFG["use_amp"]
        and torch.cuda.is_available()
    )

    for features, labels in tqdm(
        loader,
        desc=description,
        leave=False
    ):

        features = features.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=amp_enabled
        ):

            output = model(
                features,
                return_features=True
            )

            logits = output["logits"]

            loss = criterion(
                logits,
                labels
            )

        probabilities = F.softmax(
            logits.float(),
            dim=1
        )

        predictions = probabilities.argmax(
            dim=1
        )

        bs = labels.size(0)

        total_loss += (
            loss.item() * bs
        )

        total_samples += bs

        all_true.extend(
            labels.cpu().numpy()
        )

        all_pred.extend(
            predictions.cpu().numpy()
        )

        all_prob.append(
            probabilities.cpu().numpy()
        )

        all_features.append(
            output["features"].float().cpu().numpy()
        )

    y = np.asarray(all_true)
    p = np.asarray(all_pred)
    prob = np.concatenate(all_prob, axis=0)
    features = np.concatenate(
        all_features,
        axis=0
    )

    metrics = calculate_metrics(y, p)

    metrics["loss"] = (
        total_loss / max(total_samples, 1)
    )

    try:

        y_bin = label_binarize(
            y,
            classes=np.arange(NUM_CLASSES)
        )

        metrics["roc_auc"] = roc_auc_score(
            y_bin,
            prob,
            average="macro",
            multi_class="ovr"
        )

    except Exception:

        metrics["roc_auc"] = np.nan

    return (
        metrics,
        y,
        p,
        prob,
        features
    )


# ================================================================
# 19. TRAINING LOOP
# ================================================================

history = {
    "epoch": [],
    "train_loss": [],
    "val_loss": [],
    "train_accuracy": [],
    "val_accuracy": [],
    "train_precision": [],
    "val_precision": [],
    "train_recall": [],
    "val_recall": [],
    "train_macro_f1": [],
    "val_macro_f1": [],
    "val_mcc": [],
    "val_roc_auc": [],
    "learning_rate": [],
    "epoch_time_min": []
}

best_val_f1 = -np.inf
best_state = None

print("\n" + "=" * 95)
print("START FAST FUSION TRAINING")
print("=" * 95)
print("Epochs:", CFG["epochs"])
print("Training batch:", CFG["fusion_batch_size"])
print("Train samples including augmented views:", len(X_train))
print("CNN forward passes are now CACHED.")
print("=" * 95)

for epoch in range(CFG["epochs"]):

    epoch_start = time.time()

    print(
        f"\n{'='*95}\n"
        f"EPOCH {epoch+1}/{CFG['epochs']}\n"
        f"{'='*95}"
    )

    train_metrics = train_one_epoch()

    val_metrics, _, _, _, _ = evaluate(
        val_loader,
        "Validation"
    )

    scheduler.step()

    epoch_time = (
        time.time() - epoch_start
    ) / 60.0

    lr_now = optimizer.param_groups[0]["lr"]

    history["epoch"].append(epoch + 1)

    history["train_loss"].append(
        train_metrics["loss"]
    )

    history["val_loss"].append(
        val_metrics["loss"]
    )

    history["train_accuracy"].append(
        train_metrics["accuracy"]
    )

    history["val_accuracy"].append(
        val_metrics["accuracy"]
    )

    history["train_precision"].append(
        train_metrics["precision"]
    )

    history["val_precision"].append(
        val_metrics["precision"]
    )

    history["train_recall"].append(
        train_metrics["recall"]
    )

    history["val_recall"].append(
        val_metrics["recall"]
    )

    history["train_macro_f1"].append(
        train_metrics["macro_f1"]
    )

    history["val_macro_f1"].append(
        val_metrics["macro_f1"]
    )

    history["val_mcc"].append(
        val_metrics["mcc"]
    )

    history["val_roc_auc"].append(
        val_metrics["roc_auc"]
    )

    history["learning_rate"].append(
        lr_now
    )

    history["epoch_time_min"].append(
        epoch_time
    )

    print("\nPER-EPOCH RESULTS")
    print("-" * 60)
    print(
        f"Train Loss     : "
        f"{train_metrics['loss']:.5f}"
    )

    print(
        f"Val Loss       : "
        f"{val_metrics['loss']:.5f}"
    )

    print(
        f"Train Accuracy : "
        f"{train_metrics['accuracy']:.5f}"
    )

    print(
        f"Val Accuracy   : "
        f"{val_metrics['accuracy']:.5f}"
    )

    print(
        f"Train Macro-F1 : "
        f"{train_metrics['macro_f1']:.5f}"
    )

    print(
        f"Val Macro-F1   : "
        f"{val_metrics['macro_f1']:.5f}"
    )

    print(
        f"Val Precision  : "
        f"{val_metrics['precision']:.5f}"
    )

    print(
        f"Val Recall     : "
        f"{val_metrics['recall']:.5f}"
    )

    print(
        f"Val MCC        : "
        f"{val_metrics['mcc']:.5f}"
    )

    print(
        f"Val ROC-AUC    : "
        f"{val_metrics['roc_auc']:.5f}"
    )

    print(
        f"Learning Rate  : "
        f"{lr_now:.8f}"
    )

    print(
        f"Epoch Time     : "
        f"{epoch_time:.2f} min"
    )

    if val_metrics["macro_f1"] > best_val_f1:

        best_val_f1 = (
            val_metrics["macro_f1"]
        )

        best_state = copy.deepcopy(
            model.state_dict()
        )

        torch.save(
            {
                "model_state_dict": best_state,
                "class_names": class_names,
                "config": CFG,
                "best_val_macro_f1": float(
                    best_val_f1
                )
            },
            os.path.join(
                CFG["output_dir"],
                "best_RLDAF_DualCNN_Fusion.pth"
            )
        )

        print(
            "\n*** BEST MODEL SAVED ***"
        )

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

# Restore best model.
if best_state is not None:
    model.load_state_dict(
        best_state
    )

history_df = pd.DataFrame(history)

history_df.to_csv(
    os.path.join(
        CFG["output_dir"],
        "training_history.csv"
    ),
    index=False
)

# ================================================================
# 20. FINAL TEST
# ================================================================

print("\n" + "=" * 95)
print("FINAL TEST EVALUATION")
print("=" * 95)

test_metrics, y_true, y_pred, y_prob, test_features = evaluate(
    test_loader,
    "Test"
)

for key, value in test_metrics.items():

    if isinstance(
        value,
        (float, np.floating)
    ):

        print(
            f"{key:20s}: "
            f"{value:.5f}"
        )

# ================================================================
# 21. CLASSIFICATION REPORT
# ================================================================

report = classification_report(
    y_true,
    y_pred,
    target_names=class_names,
    digits=4,
    zero_division=0
)

print("\n" + "=" * 95)
print("CLASSIFICATION REPORT")
print("=" * 95)
print(report)

with open(
    os.path.join(
        CFG["output_dir"],
        "classification_report.txt"
    ),
    "w"
) as f:
    f.write(report)

# ================================================================
# 22. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES)
)

plt.figure(figsize=(9, 7))

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title("Confusion Matrix")

plt.colorbar()

plt.xticks(
    np.arange(NUM_CLASSES),
    class_names,
    rotation=45,
    ha="right"
)

plt.yticks(
    np.arange(NUM_CLASSES),
    class_names
)

for i in range(NUM_CLASSES):
    for j in range(NUM_CLASSES):
        plt.text(
            j,
            i,
            str(cm[i, j]),
            ha="center",
            va="center"
        )

plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "confusion_matrix.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 23. NORMALIZED CONFUSION MATRIX
# ================================================================

row_sum = cm.sum(
    axis=1,
    keepdims=True
)

cm_norm = np.divide(
    cm.astype(float),
    row_sum,
    out=np.zeros_like(
        cm,
        dtype=float
    ),
    where=row_sum != 0
)

plt.figure(figsize=(9, 7))

plt.imshow(
    cm_norm,
    interpolation="nearest"
)

plt.title(
    "Normalized Confusion Matrix"
)

plt.colorbar()

plt.xticks(
    np.arange(NUM_CLASSES),
    class_names,
    rotation=45,
    ha="right"
)

plt.yticks(
    np.arange(NUM_CLASSES),
    class_names
)

for i in range(NUM_CLASSES):
    for j in range(NUM_CLASSES):
        plt.text(
            j,
            i,
            f"{cm_norm[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "normalized_confusion_matrix.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 24. ROC-AUC CURVES
# ================================================================

y_bin = label_binarize(
    y_true,
    classes=np.arange(NUM_CLASSES)
)

plt.figure(figsize=(10, 8))

for i, class_name in enumerate(class_names):

    try:

        fpr, tpr, _ = roc_curve(
            y_bin[:, i],
            y_prob[:, i]
        )

        auc_score = roc_auc_score(
            y_bin[:, i],
            y_prob[:, i]
        )

        plt.plot(
            fpr,
            tpr,
            label=f"{class_name} (AUC={auc_score:.3f})"
        )

    except Exception:
        pass

plt.plot(
    [0, 1],
    [0, 1],
    "--",
    label="Random"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("One-vs-Rest ROC-AUC Curves")
plt.legend(fontsize=8)
plt.grid(alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "roc_auc_curves.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 25. PRECISION-RECALL CURVES
# ================================================================

plt.figure(figsize=(10, 8))

for i, class_name in enumerate(class_names):

    try:

        precision, recall, _ = precision_recall_curve(
            y_bin[:, i],
            y_prob[:, i]
        )

        ap = average_precision_score(
            y_bin[:, i],
            y_prob[:, i]
        )

        plt.plot(
            recall,
            precision,
            label=f"{class_name} (AP={ap:.3f})"
        )

    except Exception:
        pass

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curves")
plt.legend(fontsize=8)
plt.grid(alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "precision_recall_curves.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 26. ACCURACY CURVE
# ================================================================

epochs = history_df["epoch"]

plt.figure(figsize=(10, 6))

plt.plot(
    epochs,
    history_df["train_accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    epochs,
    history_df["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training and Validation Accuracy")
plt.legend()
plt.grid(alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "accuracy_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 27. LOSS CURVE
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    epochs,
    history_df["train_loss"],
    marker="o",
    label="Training Loss"
)

plt.plot(
    epochs,
    history_df["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")
plt.legend()
plt.grid(alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "loss_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 28. MACRO-F1 CURVE
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    epochs,
    history_df["train_macro_f1"],
    marker="o",
    label="Training Macro-F1"
)

plt.plot(
    epochs,
    history_df["val_macro_f1"],
    marker="o",
    label="Validation Macro-F1"
)

plt.xlabel("Epoch")
plt.ylabel("Macro-F1")
plt.title("Training and Validation Macro-F1")
plt.legend()
plt.grid(alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "macro_f1_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 29. LEARNING RATE CURVE
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    epochs,
    history_df["learning_rate"],
    marker="o"
)

plt.xlabel("Epoch")
plt.ylabel("Learning Rate")
plt.title("Cosine Learning Rate Schedule")
plt.grid(alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "learning_rate_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 30. PER-CLASS ANALYSIS
# ================================================================

per_precision = precision_score(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES),
    average=None,
    zero_division=0
)

per_recall = recall_score(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES),
    average=None,
    zero_division=0
)

per_f1 = f1_score(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES),
    average=None,
    zero_division=0
)

support = np.bincount(
    y_true,
    minlength=NUM_CLASSES
)

per_class_df = pd.DataFrame({
    "Class": class_names,
    "Support": support,
    "Precision": per_precision,
    "Recall": per_recall,
    "F1": per_f1
})

print("\n" + "=" * 95)
print("PER-CLASS PERFORMANCE")
print("=" * 95)
print(per_class_df.to_string(index=False))

per_class_df.to_csv(
    os.path.join(
        CFG["output_dir"],
        "per_class_metrics.csv"
    ),
    index=False
)

x = np.arange(NUM_CLASSES)
width = 0.25

plt.figure(figsize=(12, 6))

plt.bar(
    x - width,
    per_precision,
    width,
    label="Precision"
)

plt.bar(
    x,
    per_recall,
    width,
    label="Recall"
)

plt.bar(
    x + width,
    per_f1,
    width,
    label="F1"
)

plt.xticks(
    x,
    class_names,
    rotation=45,
    ha="right"
)

plt.ylim(0, 1.05)
plt.ylabel("Score")
plt.title("Per-Class Classification Performance")
plt.legend()
plt.grid(axis="y", alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "per_class_performance.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 31. BOOTSTRAP 95% CONFIDENCE INTERVALS
# ================================================================

def bootstrap_metric(
    y,
    p,
    metric_function,
    iterations=1000,
    seed=42
):

    rng = np.random.default_rng(seed)

    values = []

    n = len(y)

    for _ in range(iterations):

        indices = rng.integers(
            0,
            n,
            size=n
        )

        try:
            value = metric_function(
                y[indices],
                p[indices]
            )

            values.append(value)

        except Exception:
            continue

    values = np.asarray(values)

    if len(values) == 0:
        return np.nan, np.nan, np.nan

    return (
        float(np.mean(values)),
        float(np.percentile(values, 2.5)),
        float(np.percentile(values, 97.5))
    )


bootstrap_functions = {

    "Accuracy": lambda y, p:
        accuracy_score(y, p),

    "Macro Precision": lambda y, p:
        precision_score(
            y,
            p,
            average="macro",
            zero_division=0
        ),

    "Macro Recall": lambda y, p:
        recall_score(
            y,
            p,
            average="macro",
            zero_division=0
        ),

    "Macro F1": lambda y, p:
        f1_score(
            y,
            p,
            average="macro",
            zero_division=0
        ),

    "MCC": lambda y, p:
        matthews_corrcoef(y, p)
}

bootstrap_rows = []

for name, fn in bootstrap_functions.items():

    mean_value, lower, upper = bootstrap_metric(
        y_true,
        y_pred,
        fn,
        CFG["bootstrap_iterations"],
        SEED
    )

    bootstrap_rows.append({
        "Metric": name,
        "Mean": mean_value,
        "95% CI Lower": lower,
        "95% CI Upper": upper
    })

bootstrap_df = pd.DataFrame(
    bootstrap_rows
)

print("\n" + "=" * 95)
print("BOOTSTRAP 95% CONFIDENCE INTERVALS")
print("=" * 95)
print(bootstrap_df.to_string(index=False))

bootstrap_df.to_csv(
    os.path.join(
        CFG["output_dir"],
        "bootstrap_95CI.csv"
    ),
    index=False
)

# Statistical validation figure.
positions = np.arange(
    len(bootstrap_df)
)

means = bootstrap_df["Mean"].values

lower_error = (
    means
    - bootstrap_df["95% CI Lower"].values
)

upper_error = (
    bootstrap_df["95% CI Upper"].values
    - means
)

plt.figure(figsize=(10, 6))

plt.errorbar(
    positions,
    means,
    yerr=[
        lower_error,
        upper_error
    ],
    fmt="o",
    capsize=5
)

plt.xticks(
    positions,
    bootstrap_df["Metric"],
    rotation=30,
    ha="right"
)

plt.ylim(0, 1.05)
plt.ylabel("Score")
plt.title(
    "Bootstrap 95% Confidence Intervals"
)
plt.grid(axis="y", alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "statistical_validation.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 32. CALIBRATION / ECE
# ================================================================

def expected_calibration_error(
    y,
    probabilities,
    bins=15
):

    confidence = probabilities.max(
        axis=1
    )

    predictions = probabilities.argmax(
        axis=1
    )

    correctness = (
        predictions == y
    ).astype(float)

    edges = np.linspace(
        0.0,
        1.0,
        bins + 1
    )

    ece = 0.0
    rows = []

    for i in range(bins):

        if i == bins - 1:

            mask = (
                (confidence >= edges[i])
                &
                (confidence <= edges[i + 1])
            )

        else:

            mask = (
                (confidence >= edges[i])
                &
                (confidence < edges[i + 1])
            )

        if mask.sum() == 0:
            continue

        mean_conf = confidence[mask].mean()
        mean_acc = correctness[mask].mean()

        ece += (
            mask.sum()
            / len(y)
            * abs(mean_acc - mean_conf)
        )

        rows.append({
            "Mean Confidence": mean_conf,
            "Accuracy": mean_acc,
            "Count": int(mask.sum())
        })

    return ece, pd.DataFrame(rows)


ece, calibration_df = expected_calibration_error(
    y_true,
    y_prob
)

print("\nExpected Calibration Error:", f"{ece:.5f}")

calibration_df.to_csv(
    os.path.join(
        CFG["output_dir"],
        "calibration_data.csv"
    ),
    index=False
)

plt.figure(figsize=(8, 7))

plt.plot(
    [0, 1],
    [0, 1],
    "--",
    label="Perfect Calibration"
)

if len(calibration_df):

    plt.plot(
        calibration_df["Mean Confidence"],
        calibration_df["Accuracy"],
        "o-",
        label="RLDAF-Net"
    )

plt.xlabel("Mean Confidence")
plt.ylabel("Accuracy")

plt.title(
    f"Reliability Diagram (ECE={ece:.4f})"
)

plt.legend()
plt.grid(alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "reliability_diagram.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 33. CONFIDENCE DISTRIBUTION
# ================================================================

confidence = y_prob.max(
    axis=1
)

correct_mask = (
    y_true == y_pred
)

plt.figure(figsize=(10, 6))

if correct_mask.sum():

    plt.hist(
        confidence[correct_mask],
        bins=20,
        alpha=0.60,
        label="Correct"
    )

if (~correct_mask).sum():

    plt.hist(
        confidence[~correct_mask],
        bins=20,
        alpha=0.60,
        label="Incorrect"
    )

plt.xlabel("Prediction Confidence")
plt.ylabel("Number of Images")
plt.title("Prediction Confidence Distribution")
plt.legend()
plt.grid(alpha=0.20)
plt.tight_layout()

plt.savefig(
    os.path.join(
        CFG["output_dir"],
        "confidence_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 34. FEATURE SEPARABILITY USING t-SNE
# ================================================================

max_tsne_samples = min(
    2000,
    len(test_features)
)

rng = np.random.default_rng(
    SEED
)

if len(test_features) > max_tsne_samples:

    tsne_indices = rng.choice(
        len(test_features),
        max_tsne_samples,
        replace=False
    )

else:

    tsne_indices = np.arange(
        len(test_features)
    )

if len(tsne_indices) >= 50:

    tsne_input = test_features[
        tsne_indices
    ]

    perplexity = min(
        30,
        max(
            5,
            (len(tsne_input) - 1) // 4
        )
    )

    print(
        "\nRunning t-SNE with "
        f"{len(tsne_input)} samples..."
    )

    tsne = TSNE(
        n_components=2,
        perplexity=perplexity,
        learning_rate="auto",
        init="pca",
        random_state=SEED,
        max_iter=1000
    )

    embedding_2d = tsne.fit_transform(
        tsne_input
    )

    plt.figure(figsize=(11, 8))

    for class_id, class_name in enumerate(
        class_names
    ):

        mask = (
            y_true[tsne_indices]
            == class_id
        )

        if mask.sum():

            plt.scatter(
                embedding_2d[mask, 0],
                embedding_2d[mask, 1],
                s=15,
                alpha=0.65,
                label=class_name
            )

    plt.xlabel("t-SNE Dimension 1")
    plt.ylabel("t-SNE Dimension 2")

    plt.title(
        "t-SNE Visualization of Learned Disease Features"
    )

    plt.legend(fontsize=8)
    plt.grid(alpha=0.15)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            CFG["output_dir"],
            "tsne_features.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

# ================================================================
# 35. ERROR ANALYSIS
# ================================================================

error_df = test_df.copy()

error_df["True Class"] = [
    idx_to_class[int(x)]
    for x in y_true
]

error_df["Predicted Class"] = [
    idx_to_class[int(x)]
    for x in y_pred
]

error_df["Confidence"] = confidence

error_df["Correct"] = correct_mask

errors_only = error_df[
    ~error_df["Correct"]
].copy()

print("\n" + "=" * 95)
print("ERROR ANALYSIS")
print("=" * 95)
print(
    "Incorrect predictions:",
    len(errors_only),
    "/",
    len(error_df)
)

if len(errors_only):

    error_pairs = (
        errors_only
        .groupby(
            ["True Class", "Predicted Class"]
        )
        .size()
        .reset_index(
            name="Count"
        )
        .sort_values(
            "Count",
            ascending=False
        )
    )

    print("\nMost frequent confusion pairs:")
    print(
        error_pairs.head(20).to_string(
            index=False
        )
    )

    error_pairs.to_csv(
        os.path.join(
            CFG["output_dir"],
            "error_confusion_pairs.csv"
        ),
        index=False
    )

# ================================================================
# 36. SIMPLE ROBUSTNESS ANALYSIS
# ================================================================

print("\n" + "=" * 95)
print("ROBUSTNESS ANALYSIS")
print("=" * 95)

robust_transform = A.Compose([
    A.Resize(
        height=CFG["image_size"],
        width=CFG["image_size"]
    ),

    A.GaussNoise(
        std_range=(0.04, 0.08),
        p=1.0
    ),

    A.Normalize(
        mean=MEAN,
        std=STD
    ),

    ToTensorV2()
])

robust_loader = make_loader(
    test_df,
    robust_transform,
    batch_size=CFG["feature_batch_size"],
    shuffle=False
)

# We must reconstruct the two CNNs for robustness extraction.
robust_cnn1 = create_backbone(
    "convnextv2_tiny.fcmae_ft_in22k_in1k"
)

robust_cnn2 = create_backbone(
    "tf_efficientnetv2_s.in21k_ft_in1k"
)

robust_cnn1.eval().to(device)
robust_cnn2.eval().to(device)

for p in robust_cnn1.parameters():
    p.requires_grad = False

for p in robust_cnn2.parameters():
    p.requires_grad = False

robust_features = []
robust_labels = []

with torch.inference_mode():

    for images, labels in tqdm(
        robust_loader,
        desc="Noise robustness features",
        leave=False
    ):

        images = images.to(
            device,
            non_blocking=True
        )

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=CFG["use_amp"] and torch.cuda.is_available()
        ):

            f1 = robust_cnn1(images)
            f2 = robust_cnn2(images)

        f1 = f1.float().cpu()
        f2 = f2.float().cpu()

        robust_features.append(
            torch.cat(
                [f1, f2],
                dim=1
            )
        )

        robust_labels.append(
            labels
        )

robust_features = torch.cat(
    robust_features
)

robust_labels = torch.cat(
    robust_labels
)

# Apply training feature normalization.
robust_features = (
    (
        robust_features.float()
        - feature_mean
    )
    / feature_std
).half()

robust_ds = TensorDataset(
    robust_features,
    robust_labels
)

robust_loader_cached = DataLoader(
    robust_ds,
    batch_size=CFG["fusion_batch_size"],
    shuffle=False
)

robust_metrics, _, _, _, _ = evaluate(
    robust_loader_cached,
    "Noisy Test"
)

print(
    "Clean Test Accuracy   :",
    f"{test_metrics['accuracy']:.5f}"
)

print(
    "Noisy Test Accuracy   :",
    f"{robust_metrics['accuracy']:.5f}"
)

print(
    "Clean Test Macro-F1   :",
    f"{test_metrics['macro_f1']:.5f}"
)

print(
    "Noisy Test Macro-F1   :",
    f"{robust_metrics['macro_f1']:.5f}"
)

# Delete robustness backbones.
del robust_cnn1
del robust_cnn2
gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()

# ================================================================
# 37. TEST PREDICTIONS CSV
# ================================================================

prediction_df = test_df.copy()

prediction_df["True Class"] = [
    idx_to_class[int(x)]
    for x in y_true
]

prediction_df["Predicted Class"] = [
    idx_to_class[int(x)]
    for x in y_pred
]

prediction_df["Confidence"] = confidence

prediction_df["Correct"] = correct_mask

for i, class_name in enumerate(
    class_names
):

    safe_name = (
        class_name
        .replace(" ", "_")
        .replace("/", "_")
    )

    prediction_df[
        f"Probability_{safe_name}"
    ] = y_prob[:, i]

prediction_df.to_csv(
    os.path.join(
        CFG["output_dir"],
        "test_predictions.csv"
    ),
    index=False
)

# ================================================================
# 38. FINAL RESULTS TABLE
# ================================================================

final_results = pd.DataFrame({

    "Metric": [
        "Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1",
        "Weighted F1",
        "MCC",
        "ROC-AUC",
        "ECE",
        "Noisy Accuracy",
        "Noisy Macro F1"
    ],

    "Score": [
        test_metrics["accuracy"],
        test_metrics["precision"],
        test_metrics["recall"],
        test_metrics["macro_f1"],
        test_metrics["weighted_f1"],
        test_metrics["mcc"],
        test_metrics["roc_auc"],
        ece,
        robust_metrics["accuracy"],
        robust_metrics["macro_f1"]
    ]
})

print("\n" + "=" * 95)
print("FINAL RESULTS")
print("=" * 95)
print(
    final_results.to_string(
        index=False
    )
)

final_results.to_csv(
    os.path.join(
        CFG["output_dir"],
        "final_results.csv"
    ),
    index=False
)

# ================================================================
# 39. EXPERIMENT SUMMARY
# ================================================================

summary = {

    "dataset": "RiceLeafDiseaseBD",

    "dataset_root": DATA_ROOT,

    "total_valid_images": int(len(df)),

    "train_images_original": int(
        len(train_df)
    ),

    "train_feature_samples": int(
        len(X_train)
    ),

    "validation_images": int(
        len(val_df)
    ),

    "test_images": int(
        len(test_df)
    ),

    "classes": class_names,

    "architecture": (
        "ConvNeXtV2 Tiny + EfficientNetV2-S + "
        "Feature Projection + Multi-Scale Fusion + "
        "Gated Fusion + Disease-Aware Attention + "
        "Classification Head"
    ),

    "split": "70/15/15 stratified",

    "image_size": CFG["image_size"],

    "train_views": CFG["train_views"],

    "epochs": CFG["epochs"],

    "best_validation_macro_f1": float(
        best_val_f1
    ),

    "test_accuracy": float(
        test_metrics["accuracy"]
    ),

    "test_macro_f1": float(
        test_metrics["macro_f1"]
    ),

    "test_roc_auc": float(
        test_metrics["roc_auc"]
    ),

    "test_mcc": float(
        test_metrics["mcc"]
    ),

    "ece": float(ece),

    "noisy_accuracy": float(
        robust_metrics["accuracy"]
    ),

    "noisy_macro_f1": float(
        robust_metrics["macro_f1"]
    ),

    "fusion_total_parameters": int(
        total_params
    ),

    "fusion_trainable_parameters": int(
        trainable_params
    ),

    "gpu_count": int(GPU_COUNT),

    "feature_cache": cache_path
}

with open(
    os.path.join(
        CFG["output_dir"],
        "experiment_summary.json"
    ),
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=4
    )

# ================================================================
# 40. FINAL MEMORY CLEANUP
# ================================================================

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()

print("\n" + "=" * 95)
print("RLDAF-Net DUAL-CNN PIPELINE COMPLETED SUCCESSFULLY")
print("=" * 95)

print(
    "Best Validation Macro-F1:",
    f"{best_val_f1:.5f}"
)

print(
    "Test Accuracy:",
    f"{test_metrics['accuracy']:.5f}"
)

print(
    "Test Macro-F1:",
    f"{test_metrics['macro_f1']:.5f}"
)

print(
    "Test ROC-AUC:",
    f"{test_metrics['roc_auc']:.5f}"
)

print(
    "Results directory:",
    CFG["output_dir"]
)

print("=" * 95)
```

```python
# ================================================================
# RLDAF-Net | MEMORY-SAFE FAST END-TO-END PIPELINE
# Rice Leaf Disease Classification | Kaggle | 2 x T4
#
# IMPORTANT:
# Run this entire code as ONE Kaggle notebook cell.
#
# Dataset expected:
# /kaggle/input/.../RiceLeafDiseaseBD/Original images/
#
# Main design:
#   1. Validate images
#   2. Stratified 70/15/15 split BEFORE augmentation
#   3. Train augmentation only
#   4. Dual CNN feature extraction:
#        ConvNeXtV2 Tiny
#        EfficientNetV2-S
#   5. Feature caching to CPU RAM / disk
#   6. Lightweight gated fusion classifier
#   7. 10 epochs on cached features
#   8. Optional safe fine-tuning is NOT used by default
#      because the priority is Kaggle stability and speed.
#   9. Complete statistical and visual evaluation
#
# Outputs:
#   /kaggle/working/RLDAF_SAFE_RESULTS/
#
# Includes:
#   Accuracy
#   Loss
#   Macro-F1
#   Classification report
#   Confusion matrix
#   Normalized confusion matrix
#   ROC-AUC
#   Precision-Recall curves
#   Training curves
#   Bootstrap 95% CI
#   Calibration / ECE
#   Reliability diagram
#   Confidence distribution
#   Per-class analysis
#   Error analysis
#   t-SNE feature visualization
#   Saved predictions
#   Saved model
#
# NOTE:
# This is intentionally designed to avoid the previous kernel OOM.
# ================================================================


# ================================================================
# 0. PACKAGE SETUP
# ================================================================
import os
import sys
import gc
import json
import time
import random
import warnings
import subprocess
import pkgutil

warnings.filterwarnings("ignore")


def ensure_package(import_name, package_name):
    if pkgutil.find_loader(import_name) is None:
        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            package_name
        ])


ensure_package("timm", "timm")
ensure_package("albumentations", "albumentations")
ensure_package("sklearn", "scikit-learn")


# ================================================================
# 1. IMPORTS
# ================================================================
import numpy as np
import pandas as pd

from PIL import Image, ImageFile

ImageFile.LOAD_TRUNCATED_IMAGES = True

import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader

import timm
import albumentations as A

from albumentations.pytorch import ToTensorV2

from tqdm.auto import tqdm

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve,
    precision_recall_curve,
    average_precision_score
)

from sklearn.preprocessing import label_binarize

from sklearn.manifold import TSNE


# ================================================================
# 2. CONFIGURATION
# ================================================================
SEED = 42

IMAGE_SIZE = 224

BATCH_SIZE_CNN = 8

BATCH_SIZE_FUSION = 256

EPOCHS = 10

NUM_WORKERS = 1

PIN_MEMORY = True

AMP_ENABLED = True

FUSION_LR = 3e-4

WEIGHT_DECAY = 1e-4

DROPOUT = 0.30

LABEL_SMOOTHING = 0.05

GRAD_CLIP = 1.0

BOOTSTRAP_N = 1000

RESULT_DIR = (
    "/kaggle/working/"
    "RLDAF_SAFE_RESULTS"
)

CACHE_DIR = os.path.join(
    RESULT_DIR,
    "feature_cache"
)

os.makedirs(
    RESULT_DIR,
    exist_ok=True
)

os.makedirs(
    CACHE_DIR,
    exist_ok=True
)


# ================================================================
# 3. SEED
# ================================================================
def seed_everything(seed=42):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    torch.cuda.manual_seed_all(seed)

    os.environ[
        "PYTHONHASHSEED"
    ] = str(seed)

    # Benchmarking is faster and safer for
    # fixed-size 224x224 CNN input.
    torch.backends.cudnn.benchmark = True

    torch.backends.cudnn.deterministic = False


seed_everything(SEED)


# ================================================================
# 4. DEVICE / GPU INFORMATION
# ================================================================
device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

gpu_count = (
    torch.cuda.device_count()
    if torch.cuda.is_available()
    else 0
)

print("=" * 100)
print("RLDAF-Net | MEMORY-SAFE FAST PIPELINE")
print("=" * 100)

print(
    "PyTorch:",
    torch.__version__
)

print(
    "Device:",
    device
)

print(
    "GPU count:",
    gpu_count
)

for gpu_id in range(
    gpu_count
):

    print(
        f"GPU {gpu_id}:",
        torch.cuda.get_device_name(
            gpu_id
        )
    )

print("=" * 100)


# ================================================================
# 5. DATASET ROOT DISCOVERY
# ================================================================
candidate_paths = [

    (
        "/kaggle/input/datasets/"
        "pranzalpranmoni/"
        "riceleaf-disease-detection/"
        "RiceLeafDiseaseBD/"
        "Original images"
    ),

    (
        "/kaggle/input/"
        "riceleaf-disease-detection/"
        "RiceLeafDiseaseBD/"
        "Original images"
    ),

    (
        "/kaggle/input/"
        "RiceLeafDiseaseBD/"
        "Original images"
    )

]


DATA_ROOT = None


for path in candidate_paths:

    if os.path.isdir(path):

        DATA_ROOT = path

        break


if DATA_ROOT is None:

    print(
        "Searching /kaggle/input recursively..."
    )

    for root, dirs, files in os.walk(
        "/kaggle/input"
    ):

        if (
            os.path.basename(
                root
            ).lower().strip()
            == "original images"
        ):

            DATA_ROOT = root

            break


if DATA_ROOT is None:

    raise FileNotFoundError(
        "Could not locate the "
        "'Original images' folder."
    )


print(
    "\nDataset root:"
)

print(DATA_ROOT)


# ================================================================
# 6. DISCOVER CLASSES / IMAGES
# ================================================================
VALID_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
)


class_names = sorted([

    name

    for name in os.listdir(
        DATA_ROOT
    )

    if os.path.isdir(
        os.path.join(
            DATA_ROOT,
            name
        )
    )

])


if len(class_names) < 2:

    raise RuntimeError(
        "Less than two classes found."
    )


class_to_idx = {

    name: index

    for index, name
    in enumerate(class_names)

}


idx_to_class = {

    index: name

    for name, index
    in class_to_idx.items()

}


records = []


for class_name in class_names:

    class_dir = os.path.join(
        DATA_ROOT,
        class_name
    )

    for root, dirs, files in os.walk(
        class_dir
    ):

        for filename in files:

            if filename.lower().endswith(
                VALID_EXTENSIONS
            ):

                records.append({

                    "path":
                        os.path.join(
                            root,
                            filename
                        ),

                    "filename":
                        filename,

                    "class_name":
                        class_name,

                    "label":
                        class_to_idx[
                            class_name
                        ]

                })


df = pd.DataFrame(
    records
)


if df.empty:

    raise RuntimeError(
        "No images found."
    )


print(
    "\nDiscovered classes:"
)

for i, class_name in enumerate(
    class_names
):

    print(
        f"{i}: {class_name}"
    )


print(
    "\nTotal images:",
    len(df)
)

print(
    "\nClass distribution:"
)

print(
    df[
        "class_name"
    ]
    .value_counts()
    .sort_index()
)


# ================================================================
# 7. IMAGE VALIDATION
# ================================================================
def check_image(path):

    try:

        with Image.open(path) as img:

            img.verify()

        return True

    except Exception:

        return False


print(
    "\nValidating images..."
)


valid = []


for path in tqdm(
    df["path"].tolist(),
    desc="Image validation"
):

    valid.append(
        check_image(path)
    )


df["valid"] = valid


invalid_count = int(
    (~df["valid"]).sum()
)


df = (
    df[
        df["valid"]
    ]
    .drop(
        columns=["valid"]
    )
    .reset_index(
        drop=True
    )
)


print(
    "Invalid images:",
    invalid_count
)

print(
    "Valid images:",
    len(df)
)


# ================================================================
# 8. STRATIFIED 70 / 15 / 15 SPLIT
# ================================================================
train_df, temp_df = train_test_split(

    df,

    test_size=0.30,

    stratify=df["label"],

    random_state=SEED

)


val_df, test_df = train_test_split(

    temp_df,

    test_size=0.50,

    stratify=temp_df["label"],

    random_state=SEED

)


train_df = train_df.reset_index(
    drop=True
)

val_df = val_df.reset_index(
    drop=True
)

test_df = test_df.reset_index(
    drop=True
)


print("\n" + "=" * 100)
print("DATA SPLIT")
print("=" * 100)

print(
    "Train:",
    len(train_df)
)

print(
    "Validation:",
    len(val_df)
)

print(
    "Test:",
    len(test_df)
)

print("\nTrain:")
print(
    train_df[
        "class_name"
    ]
    .value_counts()
    .sort_index()
)

print("\nValidation:")
print(
    val_df[
        "class_name"
    ]
    .value_counts()
    .sort_index()
)

print("\nTest:")
print(
    test_df[
        "class_name"
    ]
    .value_counts()
    .sort_index()
)


train_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "train_split.csv"
    ),
    index=False
)

val_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "validation_split.csv"
    ),
    index=False
)

test_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "test_split.csv"
    ),
    index=False
)


# ================================================================
# 9. VISUALIZE DATA DISTRIBUTION
# ================================================================
plt.figure(
    figsize=(11, 6)
)

counts = (
    df[
        "class_name"
    ]
    .value_counts()
    .reindex(class_names)
)

plt.bar(
    class_names,
    counts.values
)

plt.title(
    "Rice Leaf Disease Dataset Distribution"
)

plt.xlabel(
    "Disease Class"
)

plt.ylabel(
    "Number of Images"
)

plt.xticks(
    rotation=40,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.20
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULT_DIR,
        "dataset_class_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 10. TRANSFORMS
# ================================================================
MEAN = (
    0.485,
    0.456,
    0.406
)

STD = (
    0.229,
    0.224,
    0.225
)


train_transform = A.Compose([

    A.RandomResizedCrop(
        size=(
            IMAGE_SIZE,
            IMAGE_SIZE
        ),
        scale=(0.80, 1.0),
        ratio=(0.90, 1.10),
        p=1.0
    ),

    A.HorizontalFlip(
        p=0.50
    ),

    A.VerticalFlip(
        p=0.10
    ),

    A.Affine(
        scale=(0.92, 1.08),
        translate_percent=(
            -0.03,
            0.03
        ),
        rotate=(
            -12,
            12
        ),
        shear=(
            -3,
            3
        ),
        p=0.20
    ),

    A.OneOf([

        A.RandomBrightnessContrast(
            brightness_limit=0.12,
            contrast_limit=0.12
        ),

        A.CLAHE(
            clip_limit=2.0
        ),

        A.HueSaturationValue(
            hue_shift_limit=5,
            sat_shift_limit=8,
            val_shift_limit=8
        )

    ], p=0.30),

    A.OneOf([

        A.GaussianBlur(
            blur_limit=(3, 5)
        ),

        A.GaussNoise(
            std_range=(
                0.01,
                0.03
            )
        )

    ], p=0.08),

    A.Normalize(
        mean=MEAN,
        std=STD
    ),

    ToTensorV2()

])


eval_transform = A.Compose([

    A.Resize(
        height=IMAGE_SIZE,
        width=IMAGE_SIZE
    ),

    A.Normalize(
        mean=MEAN,
        std=STD
    ),

    ToTensorV2()

])


# ================================================================
# 11. DATASET
# ================================================================
class RiceDataset(
    Dataset
):

    def __init__(
        self,
        dataframe,
        transform
    ):

        self.df = dataframe.reset_index(
            drop=True
        )

        self.transform = transform


    def __len__(self):

        return len(
            self.df
        )


    def __getitem__(
        self,
        index
    ):

        row = self.df.iloc[
            index
        ]

        try:

            image = np.asarray(

                Image.open(
                    row["path"]
                ).convert("RGB")

            )

        except Exception:

            image = np.zeros(
                (
                    IMAGE_SIZE,
                    IMAGE_SIZE,
                    3
                ),
                dtype=np.uint8
            )


        image = self.transform(
            image=image
        )["image"]


        label = int(
            row["label"]
        )


        return image, label


# ================================================================
# 12. LOADERS
# ================================================================
def make_loader(
    dataframe,
    transform,
    shuffle,
    batch_size
):

    dataset = RiceDataset(
        dataframe,
        transform
    )

    return DataLoader(

        dataset,

        batch_size=batch_size,

        shuffle=shuffle,

        num_workers=NUM_WORKERS,

        pin_memory=(
            PIN_MEMORY
            and
            torch.cuda.is_available()
        ),

        persistent_workers=False

    )


train_loader = make_loader(
    train_df,
    train_transform,
    True,
    BATCH_SIZE_CNN
)

val_loader = make_loader(
    val_df,
    eval_transform,
    False,
    BATCH_SIZE_CNN
)

test_loader = make_loader(
    test_df,
    eval_transform,
    False,
    BATCH_SIZE_CNN
)


print(
    "\nDataLoaders created."
)

print(
    "Train batches:",
    len(train_loader)
)

print(
    "Validation batches:",
    len(val_loader)
)

print(
    "Test batches:",
    len(test_loader)
)


# ================================================================
# 13. CLASS WEIGHTS
# ================================================================
class_weights_np = compute_class_weight(

    class_weight="balanced",

    classes=np.arange(
        len(class_names)
    ),

    y=train_df[
        "label"
    ].values

)


print(
    "\nClass weights:"
)

for i, name in enumerate(
    class_names
):

    print(
        f"{name:20s}: "
        f"{class_weights_np[i]:.4f}"
    )


# ================================================================
# 14. CNN LOADING
# ================================================================
print("\n" + "=" * 100)
print("LOADING PRETRAINED CNNs")
print("=" * 100)


def safe_create_backbone(
    model_name
):

    print(
        "\nLoading:",
        model_name
    )

    try:

        model = timm.create_model(

            model_name,

            pretrained=True,

            num_classes=0,

            global_pool="avg"

        )

        print(
            "Pretrained weights loaded."
        )

    except Exception as error:

        print(
            "Pretrained weight loading "
            "failed."
        )

        print(
            str(error)[:500]
        )

        print(
            "Falling back to pretrained=False."
        )

        model = timm.create_model(

            model_name,

            pretrained=False,

            num_classes=0,

            global_pool="avg"

        )

    return model


cnn1 = safe_create_backbone(
    "convnextv2_tiny.fcmae_ft_in22k_in1k"
)

cnn2 = safe_create_backbone(
    "tf_efficientnetv2_s.in21k_ft_in1k"
)


cnn1 = cnn1.to(device)
cnn2 = cnn2.to(device)


# IMPORTANT:
# Feature extraction is inference-only.
# No DataParallel is needed here.
# This avoids duplicate full CNN replicas and reduces OOM risk.


cnn1.eval()
cnn2.eval()


for parameter in cnn1.parameters():

    parameter.requires_grad = False


for parameter in cnn2.parameters():

    parameter.requires_grad = False


CNN1_DIM = cnn1.num_features
CNN2_DIM = cnn2.num_features


print(
    "\nConvNeXtV2 feature dimension:",
    CNN1_DIM
)

print(
    "EfficientNetV2 feature dimension:",
    CNN2_DIM
)


# ================================================================
# 15. FEATURE EXTRACTION FUNCTION
# ================================================================
@torch.inference_mode()
def extract_features(
    loader,
    loader_name,
    save_path
):

    # Cache exists:
    # load instead of recalculating.
    if os.path.exists(
        save_path
    ):

        print(
            f"\nLoading cached features: "
            f"{save_path}"
        )

        cached = torch.load(

            save_path,

            map_location="cpu",

            weights_only=False

        )

        return (

            cached["f1"],

            cached["f2"],

            cached["labels"]

        )


    f1_list = []
    f2_list = []
    label_list = []


    amp_ok = (
        AMP_ENABLED
        and
        torch.cuda.is_available()
    )


    print(
        f"\nExtracting features: "
        f"{loader_name}"
    )


    for images, labels in tqdm(

        loader,

        desc=loader_name

    ):

        images = images.to(

            device,

            non_blocking=True

        )


        with torch.autocast(

            device_type="cuda",

            dtype=torch.float16,

            enabled=amp_ok

        ):

            features1 = cnn1(
                images
            )

            features2 = cnn2(
                images
            )


        # Move immediately to CPU.
        # This is critical for GPU memory safety.
        f1_list.append(

            features1.float()
            .cpu()

        )

        f2_list.append(

            features2.float()
            .cpu()

        )

        label_list.append(
            labels.clone()
        )


        del (
            images,
            features1,
            features2
        )


        if torch.cuda.is_available():

            torch.cuda.empty_cache()


    f1 = torch.cat(
        f1_list,
        dim=0
    )

    f2 = torch.cat(
        f2_list,
        dim=0
    )

    labels = torch.cat(
        label_list,
        dim=0
    )


    torch.save(

        {

            "f1":
                f1,

            "f2":
                f2,

            "labels":
                labels

        },

        save_path

    )


    print(
        "Saved cache:",
        save_path
    )

    print(
        "Feature shapes:",
        tuple(f1.shape),
        tuple(f2.shape)
    )


    del (
        f1_list,
        f2_list,
        label_list
    )

    gc.collect()

    return (
        f1,
        f2,
        labels
    )


# ================================================================
# 16. FEATURE CACHE
# ================================================================
train_cache = os.path.join(
    CACHE_DIR,
    "train_features.pt"
)

val_cache = os.path.join(
    CACHE_DIR,
    "val_features.pt"
)

test_cache = os.path.join(
    CACHE_DIR,
    "test_features.pt"
)


train_f1, train_f2, train_labels = (
    extract_features(
        train_loader,
        "TRAIN CNN FEATURES",
        train_cache
    )
)


# Remove training loader from memory
# after extraction.
del train_loader
gc.collect()


val_f1, val_f2, val_labels = (
    extract_features(
        val_loader,
        "VALIDATION CNN FEATURES",
        val_cache
    )
)


del val_loader
gc.collect()


test_f1, test_f2, test_labels = (
    extract_features(
        test_loader,
        "TEST CNN FEATURES",
        test_cache
    )
)


del test_loader
gc.collect()


# Release CNN GPU memory.
del cnn1
del cnn2

gc.collect()

if torch.cuda.is_available():

    torch.cuda.empty_cache()


print(
    "\nCNN feature extraction completed."
)


# ================================================================
# 17. FEATURE NORMALIZATION
# ================================================================
# Fit normalization ONLY on training features.
# This prevents validation/test leakage.


train_f1_mean = train_f1.mean(
    dim=0,
    keepdim=True
)

train_f1_std = train_f1.std(
    dim=0,
    keepdim=True
).clamp_min(1e-6)


train_f2_mean = train_f2.mean(
    dim=0,
    keepdim=True
)

train_f2_std = train_f2.std(
    dim=0,
    keepdim=True
).clamp_min(1e-6)


def normalize_features(
    f1,
    f2
):

    f1 = (
        f1
        -
        train_f1_mean
    ) / train_f1_std

    f2 = (
        f2
        -
        train_f2_mean
    ) / train_f2_std

    return (
        f1,
        f2
    )


train_f1, train_f2 = normalize_features(
    train_f1,
    train_f2
)

val_f1, val_f2 = normalize_features(
    val_f1,
    val_f2
)

test_f1, test_f2 = normalize_features(
    test_f1,
    test_f2
)


# ================================================================
# 18. FEATURE DATASET
# ================================================================
class FeatureDataset(
    Dataset
):

    def __init__(
        self,
        f1,
        f2,
        labels
    ):

        self.f1 = f1
        self.f2 = f2
        self.labels = labels


    def __len__(self):

        return len(
            self.labels
        )


    def __getitem__(
        self,
        index
    ):

        return (

            self.f1[index],

            self.f2[index],

            self.labels[index]

        )


train_feature_dataset = FeatureDataset(

    train_f1,

    train_f2,

    train_labels

)


val_feature_dataset = FeatureDataset(

    val_f1,

    val_f2,

    val_labels

)


test_feature_dataset = FeatureDataset(

    test_f1,

    test_f2,

    test_labels

)


train_feature_loader = DataLoader(

    train_feature_dataset,

    batch_size=BATCH_SIZE_FUSION,

    shuffle=True,

    num_workers=0,

    pin_memory=(
        torch.cuda.is_available()
    )

)


val_feature_loader = DataLoader(

    val_feature_dataset,

    batch_size=BATCH_SIZE_FUSION,

    shuffle=False,

    num_workers=0,

    pin_memory=(
        torch.cuda.is_available()
    )

)


test_feature_loader = DataLoader(

    test_feature_dataset,

    batch_size=BATCH_SIZE_FUSION,

    shuffle=False,

    num_workers=0,

    pin_memory=(
        torch.cuda.is_available()
    )

)


# ================================================================
# 19. LIGHTWEIGHT RLDAF FUSION MODEL
# ================================================================
class Projection(
    nn.Module
):

    def __init__(
        self,
        in_dim,
        out_dim,
        dropout=0.30
    ):

        super().__init__()

        self.net = nn.Sequential(

            nn.Linear(
                in_dim,
                out_dim
            ),

            nn.LayerNorm(
                out_dim
            ),

            nn.GELU(),

            nn.Dropout(
                dropout
            )

        )


    def forward(self, x):

        return self.net(x)


class GatedFusion(
    nn.Module
):

    def __init__(
        self,
        dim
    ):

        super().__init__()

        self.gate = nn.Sequential(

            nn.Linear(
                dim * 2,
                dim
            ),

            nn.LayerNorm(
                dim
            ),

            nn.GELU(),

            nn.Linear(
                dim,
                dim
            ),

            nn.Sigmoid()

        )


        self.mix = nn.Sequential(

            nn.Linear(
                dim * 2,
                dim
            ),

            nn.LayerNorm(
                dim
            ),

            nn.GELU(),

            nn.Linear(
                dim,
                dim
            )

        )


    def forward(
        self,
        a,
        b
    ):

        combined = torch.cat(
            [a, b],
            dim=1
        )

        gate = self.gate(
            combined
        )

        mixed = self.mix(
            combined
        )

        fused = (
            gate * a
            +
            (1.0 - gate) * b
        )

        fused = (
            fused
            +
            mixed
        )

        return fused, gate


class MultiScaleFusion(
    nn.Module
):

    def __init__(
        self,
        dim
    ):

        super().__init__()

        self.local = nn.Sequential(

            nn.Linear(
                dim,
                dim
            ),

            nn.LayerNorm(
                dim
            ),

            nn.GELU()

        )

        self.medium = nn.Sequential(

            nn.Linear(
                dim,
                dim
            ),

            nn.LayerNorm(
                dim
            ),

            nn.GELU(),

            nn.Dropout(
                0.20
            )

        )

        self.global_block = nn.Sequential(

            nn.Linear(
                dim,
                dim
            ),

            nn.LayerNorm(
                dim
            ),

            nn.GELU(),

            nn.Linear(
                dim,
                dim
            )

        )


    def forward(self, x):

        return (

            self.local(x)

            +

            self.medium(x)

            +

            self.global_block(x)

        ) / 3.0


class DiseaseAttention(
    nn.Module
):

    def __init__(
        self,
        dim
    ):

        super().__init__()

        hidden = max(
            64,
            dim // 4
        )

        self.attention = nn.Sequential(

            nn.LayerNorm(
                dim
            ),

            nn.Linear(
                dim,
                hidden
            ),

            nn.GELU(),

            nn.Linear(
                hidden,
                dim
            ),

            nn.Sigmoid()

        )


    def forward(self, x):

        weights = self.attention(
            x
        )

        return (
            x * weights,
            weights
        )


class RLDAF_FusionNet(
    nn.Module
):

    def __init__(
        self,
        dim1,
        dim2,
        num_classes
    ):

        super().__init__()

        fusion_dim = 512


        self.proj1 = Projection(

            dim1,

            fusion_dim,

            DROPOUT

        )


        self.proj2 = Projection(

            dim2,

            fusion_dim,

            DROPOUT

        )


        self.gated = GatedFusion(
            fusion_dim
        )


        self.multiscale = MultiScaleFusion(
            fusion_dim
        )


        self.attention = DiseaseAttention(
            fusion_dim
        )


        self.norm = nn.LayerNorm(
            fusion_dim
        )


        self.embedding = nn.Sequential(

            nn.Linear(
                fusion_dim,
                256
            ),

            nn.GELU(),

            nn.Linear(
                256,
                128
            )

        )


        self.classifier = nn.Sequential(

            nn.Linear(
                fusion_dim,
                256
            ),

            nn.LayerNorm(
                256
            ),

            nn.GELU(),

            nn.Dropout(
                DROPOUT
            ),

            nn.Linear(
                256,
                128
            ),

            nn.GELU(),

            nn.Dropout(
                DROPOUT
            ),

            nn.Linear(
                128,
                num_classes
            )

        )


    def forward(
        self,
        f1,
        f2,
        return_features=False
    ):

        p1 = self.proj1(
            f1
        )

        p2 = self.proj2(
            f2
        )


        fused, gate = self.gated(

            p1,

            p2

        )


        multiscale = self.multiscale(
            fused
        )


        fused = (
            fused
            +
            multiscale
        )


        fused, attention = (
            self.attention(
                fused
            )
        )


        fused = self.norm(
            fused
        )


        logits = self.classifier(
            fused
        )


        embedding = F.normalize(

            self.embedding(
                fused
            ),

            dim=1

        )


        if return_features:

            return {

                "logits":
                    logits,

                "features":
                    fused,

                "embedding":
                    embedding,

                "gate":
                    gate,

                "attention":
                    attention

            }


        return logits


fusion_model = RLDAF_FusionNet(

    CNN1_DIM,

    CNN2_DIM,

    len(class_names)

).to(device)


# DataParallel is safe here because
# the fusion model is small.
if gpu_count > 1:

    fusion_model = nn.DataParallel(
        fusion_model
    )


print(
    "\nFusion model parameters:",
    sum(
        p.numel()
        for p in fusion_model.parameters()
    )
    / 1e6,
    "M"
)


# ================================================================
# 20. LOSS / OPTIMIZER
# ================================================================
class_weights_tensor = torch.tensor(

    class_weights_np,

    dtype=torch.float32,

    device=device

)


criterion = nn.CrossEntropyLoss(

    weight=class_weights_tensor,

    label_smoothing=LABEL_SMOOTHING

)


optimizer = torch.optim.AdamW(

    fusion_model.parameters(),

    lr=FUSION_LR,

    weight_decay=WEIGHT_DECAY

)


scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(

    optimizer,

    T_max=EPOCHS

)


try:

    scaler = torch.amp.GradScaler(

        "cuda",

        enabled=(
            AMP_ENABLED
            and
            torch.cuda.is_available()
        )

    )

except Exception:

    scaler = torch.cuda.amp.GradScaler(

        enabled=(
            AMP_ENABLED
            and
            torch.cuda.is_available()
        )

    )


# ================================================================
# 21. METRIC FUNCTION
# ================================================================
def calculate_metrics(
    y_true,
    y_pred
):

    return {

        "accuracy":
            accuracy_score(
                y_true,
                y_pred
            ),

        "balanced_accuracy":
            balanced_accuracy_score(
                y_true,
                y_pred
            ),

        "precision":
            precision_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "recall":
            recall_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "macro_f1":
            f1_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "weighted_f1":
            f1_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            ),

        "mcc":
            matthews_corrcoef(
                y_true,
                y_pred
            )

    }


# ================================================================
# 22. TRAIN EPOCH
# ================================================================
def train_epoch():

    fusion_model.train()


    total_loss = 0.0

    total_samples = 0

    all_true = []

    all_pred = []


    amp_ok = (
        AMP_ENABLED
        and
        torch.cuda.is_available()
    )


    progress = tqdm(

        train_feature_loader,

        desc="Training",

        leave=False

    )


    for f1, f2, labels in progress:

        f1 = f1.to(
            device,
            non_blocking=True
        )

        f2 = f2.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )


        optimizer.zero_grad(
            set_to_none=True
        )


        with torch.autocast(

            device_type="cuda",

            dtype=torch.float16,

            enabled=amp_ok

        ):

            logits = fusion_model(

                f1,

                f2

            )

            loss = criterion(

                logits,

                labels

            )


        if amp_ok:

            scaler.scale(
                loss
            ).backward()

            scaler.unscale_(
                optimizer
            )

            torch.nn.utils.clip_grad_norm_(

                fusion_model.parameters(),

                GRAD_CLIP

            )

            scaler.step(
                optimizer
            )

            scaler.update()

        else:

            loss.backward()

            torch.nn.utils.clip_grad_norm_(

                fusion_model.parameters(),

                GRAD_CLIP

            )

            optimizer.step()


        predictions = logits.argmax(
            dim=1
        )


        bs = labels.size(0)


        total_loss += (
            loss.item()
            *
            bs
        )


        total_samples += bs


        all_true.extend(

            labels.detach()
            .cpu()
            .numpy()
            .tolist()

        )


        all_pred.extend(

            predictions.detach()
            .cpu()
            .numpy()
            .tolist()

        )


        progress.set_postfix({

            "loss":
                f"{loss.item():.4f}",

            "acc":
                f"{accuracy_score(all_true, all_pred):.4f}"

        })


    y_true_local = np.asarray(
        all_true
    )

    y_pred_local = np.asarray(
        all_pred
    )


    metrics = calculate_metrics(

        y_true_local,

        y_pred_local

    )


    metrics[
        "loss"
    ] = (

        total_loss
        /
        max(
            total_samples,
            1
        )

    )


    return metrics


# ================================================================
# 23. EVALUATION
# ================================================================
@torch.inference_mode()
def evaluate_feature_loader(
    loader,
    description
):

    fusion_model.eval()


    total_loss = 0.0

    total_samples = 0

    all_true = []

    all_pred = []

    all_prob = []

    all_features = []


    amp_ok = (
        AMP_ENABLED
        and
        torch.cuda.is_available()
    )


    for f1, f2, labels in tqdm(

        loader,

        desc=description,

        leave=False

    ):

        f1 = f1.to(
            device,
            non_blocking=True
        )

        f2 = f2.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )


        with torch.autocast(

            device_type="cuda",

            dtype=torch.float16,

            enabled=amp_ok

        ):

            output = fusion_model(

                f1,

                f2,

                return_features=True

            )


            logits = output[
                "logits"
            ]


            loss = criterion(

                logits,

                labels

            )


        probabilities = F.softmax(

            logits.float(),

            dim=1

        )


        predictions = probabilities.argmax(
            dim=1
        )


        bs = labels.size(0)


        total_loss += (
            loss.item()
            *
            bs
        )


        total_samples += bs


        all_true.extend(

            labels.cpu()
            .numpy()
            .tolist()

        )


        all_pred.extend(

            predictions.cpu()
            .numpy()
            .tolist()

        )


        all_prob.append(

            probabilities.cpu()
            .numpy()

        )


        all_features.append(

            output[
                "features"
            ]
            .float()
            .cpu()
            .numpy()

        )


    y_true_local = np.asarray(
        all_true
    )

    y_pred_local = np.asarray(
        all_pred
    )

    y_prob_local = np.concatenate(

        all_prob,

        axis=0

    )

    features_local = np.concatenate(

        all_features,

        axis=0

    )


    metrics = calculate_metrics(

        y_true_local,

        y_pred_local

    )


    metrics[
        "loss"
    ] = (

        total_loss
        /
        max(
            total_samples,
            1
        )

    )


    try:

        y_bin_local = label_binarize(

            y_true_local,

            classes=np.arange(
                len(class_names)
            )

        )


        metrics[
            "roc_auc"
        ] = roc_auc_score(

            y_bin_local,

            y_prob_local,

            average="macro",

            multi_class="ovr"

        )

    except Exception:

        metrics[
            "roc_auc"
        ] = np.nan


    return (

        metrics,

        y_true_local,

        y_pred_local,

        y_prob_local,

        features_local

    )


# ================================================================
# 24. TRAINING LOOP
# ================================================================
history = {

    "epoch": [],

    "train_loss": [],

    "val_loss": [],

    "train_accuracy": [],

    "val_accuracy": [],

    "train_macro_f1": [],

    "val_macro_f1": [],

    "train_precision": [],

    "val_precision": [],

    "train_recall": [],

    "val_recall": [],

    "val_mcc": [],

    "val_roc_auc": [],

    "learning_rate": [],

    "epoch_time_min": []

}


best_f1 = -np.inf

best_state = None


print("\n" + "=" * 100)
print("START 10-EPOCH FUSION TRAINING")
print("=" * 100)


for epoch in range(
    EPOCHS
):

    start_time = time.time()


    print(
        f"\n{'='*100}"
    )

    print(
        f"EPOCH {epoch+1}/{EPOCHS}"
    )

    print(
        f"{'='*100}"
    )


    train_metrics = train_epoch()


    (
        val_metrics,
        _,
        _,
        _,
        _
    ) = evaluate_feature_loader(

        val_feature_loader,

        "Validation"

    )


    scheduler.step()


    elapsed_min = (
        time.time()
        -
        start_time
    ) / 60.0


    lr_now = optimizer.param_groups[
        0
    ]["lr"]


    history["epoch"].append(
        epoch + 1
    )

    history["train_loss"].append(
        train_metrics["loss"]
    )

    history["val_loss"].append(
        val_metrics["loss"]
    )

    history["train_accuracy"].append(
        train_metrics["accuracy"]
    )

    history["val_accuracy"].append(
        val_metrics["accuracy"]
    )

    history["train_macro_f1"].append(
        train_metrics["macro_f1"]
    )

    history["val_macro_f1"].append(
        val_metrics["macro_f1"]
    )

    history["train_precision"].append(
        train_metrics["precision"]
    )

    history["val_precision"].append(
        val_metrics["precision"]
    )

    history["train_recall"].append(
        train_metrics["recall"]
    )

    history["val_recall"].append(
        val_metrics["recall"]
    )

    history["val_mcc"].append(
        val_metrics["mcc"]
    )

    history["val_roc_auc"].append(
        val_metrics["roc_auc"]
    )

    history["learning_rate"].append(
        lr_now
    )

    history["epoch_time_min"].append(
        elapsed_min
    )


    print(
        "\nTrain Loss     :",
        f"{train_metrics['loss']:.5f}"
    )

    print(
        "Val Loss       :",
        f"{val_metrics['loss']:.5f}"
    )

    print(
        "Train Accuracy :",
        f"{train_metrics['accuracy']:.5f}"
    )

    print(
        "Val Accuracy   :",
        f"{val_metrics['accuracy']:.5f}"
    )

    print(
        "Train Macro-F1 :",
        f"{train_metrics['macro_f1']:.5f}"
    )

    print(
        "Val Macro-F1   :",
        f"{val_metrics['macro_f1']:.5f}"
    )

    print(
        "Val Precision  :",
        f"{val_metrics['precision']:.5f}"
    )

    print(
        "Val Recall     :",
        f"{val_metrics['recall']:.5f}"
    )

    print(
        "Val MCC        :",
        f"{val_metrics['mcc']:.5f}"
    )

    print(
        "Val ROC-AUC    :",
        f"{val_metrics['roc_auc']:.5f}"
    )

    print(
        "Learning Rate  :",
        f"{lr_now:.8f}"
    )

    print(
        "Epoch Time     :",
        f"{elapsed_min:.2f} min"
    )


    if torch.cuda.is_available():

        for gpu_id in range(
            torch.cuda.device_count()
        ):

            allocated = (

                torch.cuda.memory_allocated(
                    gpu_id
                )
                /
                1024**3

            )

            reserved = (

                torch.cuda.memory_reserved(
                    gpu_id
                )
                /
                1024**3

            )

            print(

                f"GPU {gpu_id}: "
                f"{allocated:.2f} GB allocated | "
                f"{reserved:.2f} GB reserved"

            )


    if (
        val_metrics["macro_f1"]
        >
        best_f1
    ):

        best_f1 = (
            val_metrics["macro_f1"]
        )


        best_state = {

            key: value.detach().cpu().clone()

            for key, value
            in fusion_model.state_dict().items()

        }


        torch.save(

            {

                "model_state_dict":
                    best_state,

                "class_names":
                    class_names,

                "class_to_idx":
                    class_to_idx,

                "best_val_macro_f1":
                    float(best_f1)

            },

            os.path.join(

                RESULT_DIR,

                "best_RLDAF_FusionNet.pth"

            )

        )


        print(
            "\n*** BEST MODEL SAVED ***"
        )


    gc.collect()

    if torch.cuda.is_available():

        torch.cuda.empty_cache()


# Restore best model
if best_state is not None:

    fusion_model.load_state_dict(
        best_state
    )


history_df = pd.DataFrame(
    history
)


history_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "training_history.csv"

    ),

    index=False

)


# ================================================================
# 25. TRAINING FIGURES
# ================================================================
epochs_arr = history_df[
    "epoch"
].values


plt.figure(
    figsize=(10, 6)
)

plt.plot(

    epochs_arr,

    history_df[
        "train_accuracy"
    ],

    marker="o",

    label="Train Accuracy"

)

plt.plot(

    epochs_arr,

    history_df[
        "val_accuracy"
    ],

    marker="o",

    label="Validation Accuracy"

)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()

plt.grid(
    alpha=0.20
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "accuracy_curve.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


plt.figure(
    figsize=(10, 6)
)

plt.plot(

    epochs_arr,

    history_df[
        "train_loss"
    ],

    marker="o",

    label="Train Loss"

)

plt.plot(

    epochs_arr,

    history_df[
        "val_loss"
    ],

    marker="o",

    label="Validation Loss"

)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Cross-Entropy Loss"
)

plt.title(
    "Training and Validation Loss"
)

plt.legend()

plt.grid(
    alpha=0.20
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "loss_curve.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


plt.figure(
    figsize=(10, 6)
)

plt.plot(

    epochs_arr,

    history_df[
        "train_macro_f1"
    ],

    marker="o",

    label="Train Macro-F1"

)

plt.plot(

    epochs_arr,

    history_df[
        "val_macro_f1"
    ],

    marker="o",

    label="Validation Macro-F1"

)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Macro-F1"
)

plt.title(
    "Training and Validation Macro-F1"
)

plt.legend()

plt.grid(
    alpha=0.20
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "macro_f1_curve.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ================================================================
# 26. FINAL TEST
# ================================================================
print("\n" + "=" * 100)
print("FINAL TEST EVALUATION")
print("=" * 100)


(
    test_metrics,
    y_true,
    y_pred,
    y_prob,
    test_features
) = evaluate_feature_loader(

    test_feature_loader,

    "TEST"

)


for key, value in test_metrics.items():

    print(

        f"{key:22s}: "
        f"{value:.5f}"

    )


# ================================================================
# 27. CLASSIFICATION REPORT
# ================================================================
report_text = classification_report(

    y_true,

    y_pred,

    target_names=class_names,

    digits=4,

    zero_division=0

)


print(
    "\nCLASSIFICATION REPORT"
)

print(
    "=" * 100
)

print(
    report_text
)


with open(

    os.path.join(

        RESULT_DIR,

        "classification_report.txt"

    ),

    "w"

) as file:

    file.write(
        report_text
    )


# ================================================================
# 28. CONFUSION MATRIX
# ================================================================
cm = confusion_matrix(

    y_true,

    y_pred,

    labels=np.arange(
        len(class_names)
    )

)


plt.figure(
    figsize=(9, 7)
)

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title(
    "Confusion Matrix"
)

plt.colorbar()

plt.xticks(

    np.arange(
        len(class_names)
    ),

    class_names,

    rotation=45,

    ha="right"

)

plt.yticks(

    np.arange(
        len(class_names)
    ),

    class_names

)


for i in range(
    len(class_names)
):

    for j in range(
        len(class_names)
    ):

        plt.text(

            j,

            i,

            str(
                cm[i, j]
            ),

            ha="center",

            va="center"

        )


plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "True Class"
)

plt.title(
    "Test Confusion Matrix"
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "confusion_matrix.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ================================================================
# 29. NORMALIZED CONFUSION MATRIX
# ================================================================
row_totals = cm.sum(
    axis=1,
    keepdims=True
)


normalized_cm = np.divide(

    cm.astype(float),

    row_totals,

    out=np.zeros_like(
        cm,
        dtype=float
    ),

    where=row_totals != 0

)


plt.figure(
    figsize=(9, 7)
)

plt.imshow(
    normalized_cm,
    interpolation="nearest"
)

plt.title(
    "Normalized Confusion Matrix"
)

plt.colorbar()

plt.xticks(

    np.arange(
        len(class_names)
    ),

    class_names,

    rotation=45,

    ha="right"

)

plt.yticks(

    np.arange(
        len(class_names)
    ),

    class_names

)


for i in range(
    len(class_names)
):

    for j in range(
        len(class_names)
    ):

        plt.text(

            j,

            i,

            f"{normalized_cm[i,j]:.2f}",

            ha="center",

            va="center"

        )


plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "True Class"
)

plt.title(
    "Normalized Confusion Matrix"
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "normalized_confusion_matrix.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ================================================================
# 30. ROC-AUC CURVE
# ================================================================
y_bin = label_binarize(

    y_true,

    classes=np.arange(
        len(class_names)
    )

)


plt.figure(
    figsize=(10, 8)
)


roc_auc_per_class = {}


for i, class_name in enumerate(
    class_names
):

    try:

        fpr, tpr, _ = roc_curve(

            y_bin[:, i],

            y_prob[:, i]

        )


        class_auc = roc_auc_score(

            y_bin[:, i],

            y_prob[:, i]

        )


        roc_auc_per_class[
            class_name
        ] = float(
            class_auc
        )


        plt.plot(

            fpr,

            tpr,

            label=(

                f"{class_name} "
                f"(AUC={class_auc:.3f})"

            )

        )

    except Exception:

        pass


plt.plot(

    [0, 1],

    [0, 1],

    "--",

    label="Random"

)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "One-vs-Rest ROC-AUC"
)

plt.legend(
    fontsize=8
)

plt.grid(
    alpha=0.20
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "roc_auc_curve.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ================================================================
# 31. PRECISION-RECALL CURVE
# ================================================================
plt.figure(
    figsize=(10, 8)
)


average_precision_per_class = {}


for i, class_name in enumerate(
    class_names
):

    try:

        precision, recall, _ = (
            precision_recall_curve(

                y_bin[:, i],

                y_prob[:, i]

            )
        )


        ap = average_precision_score(

            y_bin[:, i],

            y_prob[:, i]

        )


        average_precision_per_class[
            class_name
        ] = float(ap)


        plt.plot(

            recall,

            precision,

            label=(

                f"{class_name} "
                f"(AP={ap:.3f})"

            )

        )

    except Exception:

        pass


plt.xlabel(
    "Recall"
)

plt.ylabel(
    "Precision"
)

plt.title(
    "Precision-Recall Curves"
)

plt.legend(
    fontsize=8
)

plt.grid(
    alpha=0.20
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "precision_recall_curve.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ================================================================
# 32. PER-CLASS ANALYSIS
# ================================================================
precision_each = precision_score(

    y_true,

    y_pred,

    labels=np.arange(
        len(class_names)
    ),

    average=None,

    zero_division=0

)


recall_each = recall_score(

    y_true,

    y_pred,

    labels=np.arange(
        len(class_names)
    ),

    average=None,

    zero_division=0

)


f1_each = f1_score(

    y_true,

    y_pred,

    labels=np.arange(
        len(class_names)
    ),

    average=None,

    zero_division=0

)


support_each = np.bincount(

    y_true,

    minlength=len(class_names)

)


per_class_df = pd.DataFrame({

    "Class":
        class_names,

    "Support":
        support_each,

    "Precision":
        precision_each,

    "Recall":
        recall_each,

    "F1":
        f1_each

})


print(
    "\nPER-CLASS ANALYSIS"
)

print(
    per_class_df.to_string(
        index=False
    )
)


per_class_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "per_class_metrics.csv"

    ),

    index=False

)


x = np.arange(
    len(class_names)
)

width = 0.25


plt.figure(
    figsize=(12, 6)
)

plt.bar(

    x - width,

    precision_each,

    width,

    label="Precision"

)

plt.bar(

    x,

    recall_each,

    width,

    label="Recall"

)

plt.bar(

    x + width,

    f1_each,

    width,

    label="F1"

)

plt.xticks(

    x,

    class_names,

    rotation=45,

    ha="right"

)

plt.ylim(
    0,
    1.05
)

plt.ylabel(
    "Score"
)

plt.title(
    "Per-Class Performance"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.20
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "per_class_performance.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ================================================================
# 33. BOOTSTRAP 95% CI
# ================================================================
def bootstrap_ci(

    y_true_local,

    y_pred_local,

    metric_fn,

    n_boot=1000,

    seed=42

):

    rng = np.random.default_rng(
        seed
    )

    n = len(
        y_true_local
    )

    values = []


    for _ in range(
        n_boot
    ):

        indices = rng.integers(

            0,

            n,

            size=n

        )

        try:

            score = metric_fn(

                y_true_local[
                    indices
                ],

                y_pred_local[
                    indices
                ]

            )

            values.append(
                score
            )

        except Exception:

            pass


    values = np.asarray(
        values
    )


    if len(values) == 0:

        return (
            np.nan,
            np.nan,
            np.nan
        )


    return (

        float(
            np.mean(values)
        ),

        float(
            np.percentile(
                values,
                2.5
            )
        ),

        float(
            np.percentile(
                values,
                97.5
            )
        )

    )


bootstrap_metrics = {

    "Accuracy":
        lambda y, p:
            accuracy_score(
                y,
                p
            ),

    "Balanced Accuracy":
        lambda y, p:
            balanced_accuracy_score(
                y,
                p
            ),

    "Macro Precision":
        lambda y, p:
            precision_score(
                y,
                p,
                average="macro",
                zero_division=0
            ),

    "Macro Recall":
        lambda y, p:
            recall_score(
                y,
                p,
                average="macro",
                zero_division=0
            ),

    "Macro F1":
        lambda y, p:
            f1_score(
                y,
                p,
                average="macro",
                zero_division=0
            ),

    "MCC":
        lambda y, p:
            matthews_corrcoef(
                y,
                p
            )

}


bootstrap_rows = []


for metric_name, metric_fn in (
    bootstrap_metrics.items()
):

    mean_score, lower, upper = (
        bootstrap_ci(

            y_true,

            y_pred,

            metric_fn,

            BOOTSTRAP_N,

            SEED

        )
    )


    bootstrap_rows.append({

        "Metric":
            metric_name,

        "Mean":
            mean_score,

        "95% CI Lower":
            lower,

        "95% CI Upper":
            upper

    })


bootstrap_df = pd.DataFrame(
    bootstrap_rows
)


print(
    "\nBOOTSTRAP 95% CONFIDENCE INTERVALS"
)

print(
    bootstrap_df.to_string(
        index=False
    )
)


bootstrap_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "bootstrap_95CI.csv"

    ),

    index=False

)


plt.figure(
    figsize=(10, 6)
)

positions = np.arange(
    len(bootstrap_df)
)

means = bootstrap_df[
    "Mean"
].values

lower = bootstrap_df[
    "95% CI Lower"
].values

upper = bootstrap_df[
    "95% CI Upper"
].values


plt.errorbar(

    positions,

    means,

    yerr=[

        means - lower,

        upper - means

    ],

    fmt="o",

    capsize=5

)


plt.xticks(

    positions,

    bootstrap_df[
        "Metric"
    ],

    rotation=30,

    ha="right"

)

plt.ylim(
    0,
    1.05
)

plt.ylabel(
    "Score"
)

plt.title(
    "Statistical Validation: Bootstrap 95% CI"
)

plt.grid(
    axis="y",
    alpha=0.20
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "bootstrap_ci_figure.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ================================================================
# 34. CALIBRATION / ECE
# ================================================================
def calculate_ece(

    y_true_local,

    probabilities,

    bins=15

):

    confidence = probabilities.max(
        axis=1
    )

    predictions = probabilities.argmax(
        axis=1
    )

    correct = (
        predictions
        ==
        y_true_local
    ).astype(float)


    edges = np.linspace(
        0,
        1,
        bins + 1
    )


    ece = 0.0

    rows = []


    for i in range(
        bins
    ):

        if i == bins - 1:

            mask = (

                confidence >=
                edges[i]

            ) & (

                confidence <=
                edges[i + 1]

            )

        else:

            mask = (

                confidence >=
                edges[i]

            ) & (

                confidence <
                edges[i + 1]

            )


        if mask.sum() == 0:

            continue


        avg_confidence = (
            confidence[mask].mean()
        )

        avg_accuracy = (
            correct[mask].mean()
        )


        ece += (

            mask.sum()
            /
            len(y_true_local)
            *
            abs(

                avg_accuracy
                -
                avg_confidence

            )

        )


        rows.append({

            "Confidence":
                avg_confidence,

            "Accuracy":
                avg_accuracy,

            "Count":
                int(
                    mask.sum()
                )

        })


    return (

        ece,

        pd.DataFrame(rows)

    )


ece, calibration_df = calculate_ece(

    y_true,

    y_prob

)


print(
    "\nExpected Calibration Error:",
    f"{ece:.5f}"
)


calibration_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "calibration_data.csv"

    ),

    index=False

)


plt.figure(
    figsize=(8, 7)
)

plt.plot(

    [0, 1],

    [0, 1],

    "--",

    label="Perfect Calibration"

)


if len(calibration_df):

    plt.plot(

        calibration_df[
            "Confidence"
        ],

        calibration_df[
            "Accuracy"
        ],

        "o-",

        label="RLDAF-Net"

    )


plt.xlabel(
    "Mean Confidence"
)

plt.ylabel(
    "Accuracy"
)

plt.title(
    f"Reliability Diagram | ECE={ece:.4f}"
)

plt.legend()

plt.grid(
    alpha=0.20
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "reliability_diagram.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ================================================================
# 35. CONFIDENCE DISTRIBUTION
# ================================================================
confidence = y_prob.max(
    axis=1
)

correct_mask = (
    y_true == y_pred
)


plt.figure(
    figsize=(10, 6)
)


if correct_mask.sum():

    plt.hist(

        confidence[
            correct_mask
        ],

        bins=20,

        alpha=0.60,

        label="Correct"

    )


if (~correct_mask).sum():

    plt.hist(

        confidence[
            ~correct_mask
        ],

        bins=20,

        alpha=0.60,

        label="Incorrect"

    )


plt.xlabel(
    "Prediction Confidence"
)

plt.ylabel(
    "Number of Samples"
)

plt.title(
    "Prediction Confidence Distribution"
)

plt.legend()

plt.grid(
    alpha=0.20
)

plt.tight_layout()

plt.savefig(

    os.path.join(

        RESULT_DIR,

        "confidence_distribution.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ================================================================
# 36. ERROR ANALYSIS
# ================================================================
error_df = test_df.copy()


error_df[
    "True Class"
] = [

    idx_to_class[
        int(x)
    ]

    for x in y_true

]


error_df[
    "Predicted Class"
] = [

    idx_to_class[
        int(x)
    ]

    for x in y_pred

]


error_df[
    "Confidence"
] = confidence


error_df[
    "Correct"
] = correct_mask


errors_only = error_df[
    ~error_df[
        "Correct"
    ]
].copy()


print(
    "\n" +
    "=" * 100
)

print(
    "ERROR ANALYSIS"
)

print(
    "=" * 100
)

print(
    "Total test errors:",
    len(errors_only)
)


if len(errors_only):

    error_pairs = (

        errors_only

        .groupby(

            [
                "True Class",
                "Predicted Class"
            ]

        )

        .size()

        .reset_index(
            name="Count"
        )

        .sort_values(

            "Count",

            ascending=False

        )

    )


    print(
        "\nMost frequent confusion pairs:"
    )

    print(

        error_pairs
        .head(20)
        .to_string(
            index=False
        )

    )


    error_pairs.to_csv(

        os.path.join(

            RESULT_DIR,

            "error_confusion_pairs.csv"

        ),

        index=False

    )


# ================================================================
# 37. SAVE TEST PREDICTIONS
# ================================================================
prediction_df = error_df.copy()


for i, class_name in enumerate(
    class_names
):

    safe_name = (

        class_name
        .replace(
            " ",
            "_"
        )
        .replace(
            "/",
            "_"
        )

    )


    prediction_df[
        f"Probability_{safe_name}"
    ] = y_prob[:, i]


prediction_df.to_csv(

    os.path.join(

        RESULT_DIR,

        "test_predictions.csv"

    ),

    index=False

)


# ================================================================
# 38. t-SNE FEATURE VISUALIZATION
# ================================================================
max_tsne_samples = min(

    2000,

    len(test_features)

)


rng = np.random.default_rng(
    SEED
)


if len(test_features) > max_tsne_samples:

    tsne_indices = rng.choice(

        len(test_features),

        max_tsne_samples,

        replace=False

    )

else:

    tsne_indices = np.arange(
        len(test_features)
    )


if len(tsne_indices) >= 50:

    print(
        "\nRunning t-SNE..."
    )


    tsne_features = (
        test_features[
            tsne_indices
        ]
    )


    perplexity = min(

        30,

        max(

            5,

            (len(tsne_features) - 1)
            // 4

        )

    )


    tsne = TSNE(

        n_components=2,

        perplexity=perplexity,

        learning_rate="auto",

        init="pca",

        random_state=SEED,

        max_iter=750

    )


    tsne_2d = tsne.fit_transform(

        tsne_features

    )


    plt.figure(
        figsize=(11, 8)
    )


    for class_id, class_name in enumerate(
        class_names
    ):

        mask = (

            y_true[
                tsne_indices
            ]
            ==
            class_id

        )


        if mask.sum():

            plt.scatter(

                tsne_2d[
                    mask,
                    0
                ],

                tsne_2d[
                    mask,
                    1
                ],

                s=15,

                alpha=0.65,

                label=class_name

            )


    plt.xlabel(
        "t-SNE Dimension 1"
    )

    plt.ylabel(
        "t-SNE Dimension 2"
    )

    plt.title(
        "t-SNE Visualization of Fused Disease Features"
    )

    plt.legend(
        fontsize=8
    )

    plt.grid(
        alpha=0.15
    )

    plt.tight_layout()

    plt.savefig(

        os.path.join(

            RESULT_DIR,

            "tsne_feature_visualization.png"

        ),

        dpi=300,

        bbox_inches="tight"

    )

    plt.show()


# ================================================================
# 39. FINAL RESULTS TABLE
# ================================================================
final_results = pd.DataFrame({

    "Metric": [

        "Accuracy",

        "Balanced Accuracy",

        "Macro Precision",

        "Macro Recall",

        "Macro F1",

        "Weighted F1",

        "MCC",

        "ROC-AUC",

        "ECE"

    ],

    "Score": [

        test_metrics[
            "accuracy"
        ],

        test_metrics[
            "balanced_accuracy"
        ],

        test_metrics[
            "precision"
        ],

        test_metrics[
            "recall"
        ],

        test_metrics[
            "macro_f1"
        ],

        test_metrics[
            "weighted_f1"
        ],

        test_metrics[
            "mcc"
        ],

        test_metrics[
            "roc_auc"
        ],

        ece

    ]

})


print(
    "\n" +
    "=" * 100
)

print(
    "FINAL TEST RESULTS"
)

print(
    "=" * 100
)

print(

    final_results.to_string(
        index=False
    )

)


final_results.to_csv(

    os.path.join(

        RESULT_DIR,

        "final_results.csv"

    ),

    index=False

)


# ================================================================
# 40. SUMMARY JSON
# ================================================================
summary = {

    "dataset":
        "RiceLeafDiseaseBD",

    "dataset_root":
        DATA_ROOT,

    "total_images":
        int(len(df)),

    "invalid_images":
        int(invalid_count),

    "train_images":
        int(len(train_df)),

    "validation_images":
        int(len(val_df)),

    "test_images":
        int(len(test_df)),

    "classes":
        class_names,

    "split":
        "70/15/15 stratified",

    "image_size":
        IMAGE_SIZE,

    "cnn_backbones": [

        "ConvNeXtV2 Tiny",

        "EfficientNetV2-S"

    ],

    "fusion":
        (
            "Projection + "
            "Gated Fusion + "
            "Multi-Scale Fusion + "
            "Disease-Aware Attention"
        ),

    "epochs":
        EPOCHS,

    "batch_size_cnn":
        BATCH_SIZE_CNN,

    "batch_size_fusion":
        BATCH_SIZE_FUSION,

    "best_validation_macro_f1":
        float(best_f1),

    "test_accuracy":
        float(
            test_metrics[
                "accuracy"
            ]
        ),

    "test_macro_f1":
        float(
            test_metrics[
                "macro_f1"
            ]
        ),

    "test_roc_auc":
        float(
            test_metrics[
                "roc_auc"
            ]
        ),

    "test_mcc":
        float(
            test_metrics[
                "mcc"
            ]
        ),

    "test_balanced_accuracy":
        float(
            test_metrics[
                "balanced_accuracy"
            ]
        ),

    "ece":
        float(ece),

    "gpu_count":
        int(gpu_count)

}


with open(

    os.path.join(

        RESULT_DIR,

        "experiment_summary.json"

    ),

    "w"

) as file:

    json.dump(

        summary,

        file,

        indent=4

    )


# ================================================================
# 41. FINAL MODEL SAVE
# ================================================================
torch.save(

    {

        "model_state_dict":
            fusion_model.state_dict(),

        "class_names":
            class_names,

        "class_to_idx":
            class_to_idx,

        "cnn_feature_dimensions": [

            int(CNN1_DIM),

            int(CNN2_DIM)

        ],

        "image_size":
            IMAGE_SIZE,

        "best_validation_macro_f1":
            float(best_f1),

        "test_metrics":
            {
                key:
                    float(value)

                for key, value
                in test_metrics.items()

                if isinstance(
                    value,
                    (float, np.floating)
                )
            }

    },

    os.path.join(

        RESULT_DIR,

        "RLDAF_FINAL_MODEL.pth"

    )

)


# ================================================================
# 42. FINAL CLEANUP / REPORT
# ================================================================
gc.collect()

if torch.cuda.is_available():

    torch.cuda.empty_cache()


print("\n" + "=" * 100)
print("PIPELINE COMPLETED SUCCESSFULLY")
print("=" * 100)

print(
    "\nBest Validation Macro-F1:",
    f"{best_f1:.5f}"
)

print(
    "Test Accuracy:",
    f"{test_metrics['accuracy']:.5f}"
)

print(
    "Test Macro-F1:",
    f"{test_metrics['macro_f1']:.5f}"
)

print(
    "Test ROC-AUC:",
    f"{test_metrics['roc_auc']:.5f}"
)

print(
    "Test MCC:",
    f"{test_metrics['mcc']:.5f}"
)

print(
    "ECE:",
    f"{ece:.5f}"
)

print(
    "\nResults directory:"
)

print(
    RESULT_DIR
)

print(
    "\nImportant output files:"
)

for filename in sorted(
    os.listdir(
        RESULT_DIR
    )
):

    if os.path.isfile(
        os.path.join(
            RESULT_DIR,
            filename
        )
    ):

        print(
            "  ",
            filename
        )

print("=" * 100)
```

```python

# ================================================================
# DUAL-CNN RICE LEAF DISEASE CLASSIFICATION
# RLDAF-CNN: Rice Leaf Disease-Aware Dual CNN Fusion Network
#
# Dataset:
#   RiceLeafDiseaseBD / Original images
#
# Classes:
#   Blast, Brown spot, Healthy, Leaf smut, Rice Tungro, Sheath blight
#
# Main idea:
#   Original image
#        |
#   +----+--------------------+
#   |                         |
# ConvNeXtV2-Tiny        EfficientNetV2-S
#   |                         |
# Multi-scale CNN       Multi-scale CNN
#   |                         |
# CBAM attention        CBAM attention
#   |                         |
#   +------ Cross-Branch -----+
#          Attention
#              |
#       Adaptive Gate Fusion
#              |
#        Classification
#
# IMPORTANT:
#   1. Only "Original images" are used.
#   2. No bounding boxes, masks, annotations or ROI crops.
#   3. Exact duplicates are removed BEFORE splitting.
#   4. Augmentation is applied ONLY to training data.
#   5. Validation/test data are kept augmentation-free.
#   6. Split is stratified 70/15/15.
#
# Designed for Kaggle + 2 x NVIDIA T4.
# ================================================================

# ================================================================
# 0. INSTALL
# ================================================================
!pip install -q timm scikit-learn seaborn openpyxl umap-learn thop

# ================================================================
# 1. IMPORTS
# ================================================================
import os
import gc
import cv2
import json
import math
import time
import copy
import random
import hashlib
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
from PIL import Image, ImageFile

import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import timm

from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    cohen_kappa_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve,
    brier_score_loss,
)
from sklearn.preprocessing import label_binarize
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
from sklearn.calibration import calibration_curve
from sklearn.utils.class_weight import compute_class_weight

warnings.filterwarnings("ignore")
ImageFile.LOAD_TRUNCATED_IMAGES = True

# ================================================================
# 2. CONFIGURATION
# ================================================================
SEED = 42

DATASET_ROOT = Path(
    "/kaggle/input/datasets/pranzalpranmoni/"
    "riceleaf-disease-detection/RiceLeafDiseaseBD"
)

# EXACTLY the image directory requested by the project
IMAGE_ROOT = DATASET_ROOT / "Original images"

OUTPUT_DIR = Path("/kaggle/working/RLDAF_CNN_Rice_OriginalImages")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CLASS_NAMES = [
    "Blast",
    "Brown spot",
    "Healthy",
    "Leaf smut",
    "Rice Tungro",
    "Sheath blight",
]
NUM_CLASSES = len(CLASS_NAMES)

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp",
    ".tif", ".tiff", ".webp"
}

# ---------------- split ----------------
TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

# ---------------- image/training ----------------
IMAGE_SIZE = 320
BATCH_SIZE = 2
VAL_BATCH_SIZE = 8
GRAD_ACCUM_STEPS = 2
NUM_WORKERS = 2

EPOCHS = 20
PATIENCE = 5

BACKBONE_LR = 1e-5
HEAD_LR = 2e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTHING = 0.05

# loss:
# CE + auxiliary branch supervision + feature consistency
LAMBDA_AUX = 0.20
LAMBDA_CONSISTENCY = 0.05

USE_PRETRAINED = True

# ---------------- optional analysis ----------------
RUN_PCA = True
RUN_TSNE = True
RUN_UMAP = True
RUN_GRADCAM = True

# Keep expensive experiments OFF for first run.
# Turn these ON only after the main experiment finishes.
RUN_BASELINES = False
RUN_ABLATION = False
RUN_ROBUSTNESS = False
RUN_RESOLUTION = False
RUN_MULTISEED = False

MAX_FEATURE_IMAGES = 1000
MAX_CAM_IMAGES = 18
BOOTSTRAP_ROUNDS = 1000

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
N_GPUS = torch.cuda.device_count()

print("=" * 80)
print("RLDAF-CNN | DUAL CNN RICE LEAF DISEASE PIPELINE")
print("=" * 80)
print("Device:", DEVICE)
print("GPU count:", N_GPUS)
if N_GPUS:
    for i in range(N_GPUS):
        print(f"GPU {i}:", torch.cuda.get_device_name(i))
print("Image root:", IMAGE_ROOT)
print("Output:", OUTPUT_DIR)

# ================================================================
# 3. REPRODUCIBILITY
# ================================================================
def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = False
    torch.backends.cudnn.benchmark = True

seed_everything(SEED)

# ================================================================
# 4. CHECK DATASET
# ================================================================
if not IMAGE_ROOT.exists():
    raise FileNotFoundError(
        f"Original images directory was not found:\n{IMAGE_ROOT}\n"
        "Check the Kaggle dataset path."
    )

available_dirs = [p.name for p in IMAGE_ROOT.iterdir() if p.is_dir()]
print("\nDetected folders:")
for x in available_dirs:
    print("  ", x)

missing = [c for c in CLASS_NAMES if not (IMAGE_ROOT / c).exists()]
if missing:
    raise FileNotFoundError(
        f"Missing class folders inside Original images: {missing}"
    )

# ================================================================
# 5. COLLECT ONLY ORIGINAL IMAGES
# ================================================================
records = []

for class_id, class_name in enumerate(CLASS_NAMES):
    class_dir = IMAGE_ROOT / class_name

    for p in class_dir.rglob("*"):
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS:
            records.append({
                "path": str(p),
                "class_name": class_name,
                "label": class_id,
            })

df = pd.DataFrame(records)

if len(df) == 0:
    raise RuntimeError("No images were found.")

print("\nTotal images before validation:", len(df))
print(df["class_name"].value_counts().reindex(CLASS_NAMES))

# ================================================================
# 6. IMAGE VALIDATION
# ================================================================
def is_valid_image(path):
    try:
        with Image.open(path) as img:
            img.verify()
        return True
    except Exception:
        return False

valid_flags = []
for p in df["path"]:
    valid_flags.append(is_valid_image(p))

df["valid"] = valid_flags
bad_count = int((~df["valid"]).sum())

if bad_count:
    print("Invalid/corrupted images:", bad_count)
else:
    print("Invalid/corrupted images: 0")

df = df[df["valid"]].drop(columns=["valid"]).reset_index(drop=True)

# ================================================================
# 7. EXACT DUPLICATE REMOVAL BEFORE SPLIT
# ================================================================
def md5_file(path, chunk_size=1024 * 1024):
    h = hashlib.md5()
    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)
    return h.hexdigest()

print("\nCalculating exact image hashes...")
df["md5"] = [md5_file(p) for p in df["path"]]

before_dup = len(df)

# Keep the first copy of every byte-identical file.
df = (
    df.drop_duplicates(subset=["md5"], keep="first")
      .drop(columns=["md5"])
      .reset_index(drop=True)
)

print("Images before duplicate removal:", before_dup)
print("Images after duplicate removal :", len(df))
print("Exact duplicates removed       :", before_dup - len(df))

# ================================================================
# 8. DATASET DISTRIBUTION
# ================================================================
dataset_counts = (
    df["class_name"]
    .value_counts()
    .reindex(CLASS_NAMES)
    .fillna(0)
    .astype(int)
)

dataset_counts.to_csv(
    OUTPUT_DIR / "dataset_class_distribution.csv",
    header=["count"]
)

plt.figure(figsize=(10, 5))
plt.bar(dataset_counts.index, dataset_counts.values)
plt.xticks(rotation=30, ha="right")
plt.xlabel("Disease class")
plt.ylabel("Number of images")
plt.title("Rice Leaf Disease Dataset Distribution")
plt.tight_layout()
plt.savefig(
    OUTPUT_DIR / "dataset_class_distribution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ================================================================
# 9. REPRESENTATIVE IMAGE GRID
# ================================================================
def save_sample_grid(dataframe, filename="representative_images.png", n=4):
    fig, axes = plt.subplots(
        NUM_CLASSES, n,
        figsize=(3.2 * n, 3.0 * NUM_CLASSES)
    )

    for r, cls in enumerate(CLASS_NAMES):
        sub = dataframe[dataframe["class_name"] == cls]
        sub = sub.sample(
            min(n, len(sub)),
            random_state=SEED
        ).reset_index(drop=True)

        for c in range(n):
            ax = axes[r, c]
            ax.axis("off")

            if c < len(sub):
                try:
                    img = Image.open(sub.loc[c, "path"]).convert("RGB")
                    ax.imshow(img)
                    ax.set_title(cls, fontsize=9)
                except Exception:
                    pass

    plt.tight_layout()
    plt.savefig(
        OUTPUT_DIR / filename,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

save_sample_grid(df)

# ================================================================
# 10. STRATIFIED 70/15/15 SPLIT
# ================================================================
train_df, temp_df = train_test_split(
    df,
    test_size=VAL_RATIO + TEST_RATIO,
    stratify=df["label"],
    random_state=SEED
)

relative_test = TEST_RATIO / (VAL_RATIO + TEST_RATIO)

val_df, test_df = train_test_split(
    temp_df,
    test_size=relative_test,
    stratify=temp_df["label"],
    random_state=SEED
)

train_df = train_df.copy()
val_df = val_df.copy()
test_df = test_df.copy()

train_df["split"] = "train"
val_df["split"] = "validation"
test_df["split"] = "test"

master_split = pd.concat(
    [train_df, val_df, test_df],
    ignore_index=True
)

master_split.to_csv(
    OUTPUT_DIR / "master_split.csv",
    index=False
)

print("\nSplit sizes:")
print("Train      :", len(train_df))
print("Validation :", len(val_df))
print("Test       :", len(test_df))

print("\nSplit distribution:")
print(
    pd.crosstab(
        master_split["class_name"],
        master_split["split"]
    ).reindex(CLASS_NAMES)
)

# ================================================================
# 11. DATA TRANSFORMS
# ================================================================
# ImageNet normalization is used because both CNN backbones are
# initialized from ImageNet-pretrained weights.

IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE),
        interpolation=transforms.InterpolationMode.BICUBIC
    ),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.15),
    transforms.RandomRotation(20),
    transforms.ColorJitter(
        brightness=0.18,
        contrast=0.18,
        saturation=0.15,
        hue=0.03
    ),
    transforms.RandomResizedCrop(
        IMAGE_SIZE,
        scale=(0.82, 1.0),
        ratio=(0.90, 1.10),
        interpolation=transforms.InterpolationMode.BICUBIC
    ),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    transforms.RandomErasing(
        p=0.20,
        scale=(0.02, 0.10),
        ratio=(0.3, 3.3),
        value="random"
    ),
])

eval_transform = transforms.Compose([
    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE),
        interpolation=transforms.InterpolationMode.BICUBIC
    ),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

# ================================================================
# 12. DATASET
# ================================================================
class RiceLeafDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, idx):
        row = self.dataframe.iloc[idx]

        path = row["path"]
        label = int(row["label"])

        image = Image.open(path).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        return image, label, path


train_ds = RiceLeafDataset(train_df, train_transform)
val_ds = RiceLeafDataset(val_df, eval_transform)
test_ds = RiceLeafDataset(test_df, eval_transform)

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    persistent_workers=(NUM_WORKERS > 0),
    drop_last=False
)

val_loader = DataLoader(
    val_ds,
    batch_size=VAL_BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    persistent_workers=(NUM_WORKERS > 0)
)

test_loader = DataLoader(
    test_ds,
    batch_size=VAL_BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    persistent_workers=(NUM_WORKERS > 0)
)

# ================================================================
# 13. CBAM
# ================================================================
class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()

        hidden = max(channels // reduction, 8)

        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.mlp = nn.Sequential(
            nn.Conv2d(channels, hidden, 1, bias=False),
            nn.ReLU(inplace=True),
            nn.Conv2d(hidden, channels, 1, bias=False)
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_out = self.mlp(self.avg_pool(x))
        max_out = self.mlp(self.max_pool(x))
        return self.sigmoid(avg_out + max_out)


class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()

        padding = kernel_size // 2

        self.conv = nn.Conv2d(
            2, 1,
            kernel_size=kernel_size,
            padding=padding,
            bias=False
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)

        pooled = torch.cat([avg_out, max_out], dim=1)
        return self.sigmoid(self.conv(pooled))


class CBAM(nn.Module):
    def __init__(self, channels):
        super().__init__()
        self.channel = ChannelAttention(channels)
        self.spatial = SpatialAttention()

    def forward(self, x):
        x = x * self.channel(x)
        x = x * self.spatial(x)
        return x

# ================================================================
# 14. SAFE FEATURE BACKBONE
# ================================================================
def create_feature_backbone(model_name, pretrained=True):
    """
    Uses timm features_only=True without hard-coding out_indices.
    This avoids version-dependent feature index errors.
    """
    try:
        model = timm.create_model(
            model_name,
            pretrained=pretrained,
            features_only=True
        )
    except Exception as e:
        print(f"Pretrained initialization failed for {model_name}.")
        print("Trying randomly initialized backbone.")
        print("Reason:", str(e)[:300])

        model = timm.create_model(
            model_name,
            pretrained=False,
            features_only=True
        )

    channels = model.feature_info.channels()

    print(f"\n{model_name}")
    print("Feature channels:", channels)

    return model, channels

# ================================================================
# 15. MULTI-SCALE TOKEN EXTRACTOR
# ================================================================
class MultiScaleTokenExtractor(nn.Module):
    """
    Converts the final 3 CNN feature maps into three compact tokens.

    For each scale:
        CNN feature map
          -> CBAM
          -> global average pooling
          -> projection
          -> token

    The result is a sequence:
        [B, 3, D]
    """

    def __init__(self, channels, token_dim=384):
        super().__init__()

        self.indices = list(
            range(max(0, len(channels) - 3), len(channels))
        )

        selected_channels = [
            channels[i] for i in self.indices
        ]

        self.attention = nn.ModuleList([
            CBAM(c) for c in selected_channels
        ])

        self.projections = nn.ModuleList([
            nn.Sequential(
                nn.Linear(c, token_dim),
                nn.LayerNorm(token_dim),
                nn.GELU()
            )
            for c in selected_channels
        ])

    def forward(self, features):
        tokens = []

        for i, feat_idx in enumerate(self.indices):
            x = features[feat_idx]
            x = self.attention[i](x)

            x = F.adaptive_avg_pool2d(x, 1)
            x = x.flatten(1)

            x = self.projections[i](x)
            tokens.append(x)

        return torch.stack(tokens, dim=1)

# ================================================================
# 16. CROSS-BRANCH ATTENTION
# ================================================================
class BidirectionalCrossBranchAttention(nn.Module):
    """
    Branch A attends to Branch B and Branch B attends to Branch A.
    The attended token sequences are then combined.
    """

    def __init__(self, dim=384, heads=6, dropout=0.10):
        super().__init__()

        self.a_to_b = nn.MultiheadAttention(
            embed_dim=dim,
            num_heads=heads,
            dropout=dropout,
            batch_first=True
        )

        self.b_to_a = nn.MultiheadAttention(
            embed_dim=dim,
            num_heads=heads,
            dropout=dropout,
            batch_first=True
        )

        self.norm_a = nn.LayerNorm(dim)
        self.norm_b = nn.LayerNorm(dim)

        self.ffn_a = nn.Sequential(
            nn.Linear(dim, dim * 2),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(dim * 2, dim)
        )

        self.ffn_b = nn.Sequential(
            nn.Linear(dim, dim * 2),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(dim * 2, dim)
        )

    def forward(self, a, b):
        a_attn, _ = self.a_to_b(
            query=a,
            key=b,
            value=b,
            need_weights=False
        )

        b_attn, _ = self.b_to_a(
            query=b,
            key=a,
            value=a,
            need_weights=False
        )

        a = self.norm_a(a + a_attn)
        b = self.norm_b(b + b_attn)

        a = self.norm_a(a + self.ffn_a(a))
        b = self.norm_b(b + self.ffn_b(b))

        return a, b

# ================================================================
# 17. ADAPTIVE GATED FUSION
# ================================================================
class AdaptiveGatedFusion(nn.Module):
    """
    Learns sample-specific contribution of the two CNN branches.
    """

    def __init__(self, dim=384):
        super().__init__()

        self.gate = nn.Sequential(
            nn.Linear(dim * 2, dim),
            nn.GELU(),
            nn.Dropout(0.10),
            nn.Linear(dim, 2)
        )

        self.refine = nn.Sequential(
            nn.Linear(dim, dim),
            nn.LayerNorm(dim),
            nn.GELU()
        )

    def forward(self, a, b):
        a_global = a.mean(dim=1)
        b_global = b.mean(dim=1)

        gate_logits = self.gate(
            torch.cat([a_global, b_global], dim=1)
        )

        gate = torch.softmax(gate_logits, dim=1)

        fused = (
            gate[:, 0:1] * a_global +
            gate[:, 1:2] * b_global
        )

        fused = self.refine(fused)

        return fused, gate

# ================================================================
# 18. DUAL CNN MODEL
# ================================================================
class RLDAFCNN(nn.Module):
    def __init__(
        self,
        num_classes=6,
        token_dim=384,
        pretrained=True
    ):
        super().__init__()

        # --------------------------------------------------------
        # CNN Branch A: global semantic representation
        # --------------------------------------------------------
        self.branch_a, channels_a = create_feature_backbone(
            "convnextv2_tiny.fcmae_ft_in22k_in1k",
            pretrained
        )

        # --------------------------------------------------------
        # CNN Branch B: complementary fine-grained representation
        # --------------------------------------------------------
        self.branch_b, channels_b = create_feature_backbone(
            "tf_efficientnetv2_s.in21k_ft_in1k",
            pretrained
        )

        self.extractor_a = MultiScaleTokenExtractor(
            channels_a,
            token_dim
        )

        self.extractor_b = MultiScaleTokenExtractor(
            channels_b,
            token_dim
        )

        self.cross_attention = BidirectionalCrossBranchAttention(
            dim=token_dim,
            heads=6,
            dropout=0.10
        )

        self.fusion = AdaptiveGatedFusion(token_dim)

        self.classifier = nn.Sequential(
            nn.Linear(token_dim, 256),
            nn.LayerNorm(256),
            nn.GELU(),
            nn.Dropout(0.35),

            nn.Linear(256, 128),
            nn.GELU(),
            nn.Dropout(0.20),

            nn.Linear(128, num_classes)
        )

        # Auxiliary supervision encourages both CNNs
        # to learn useful disease representations.
        self.head_a = nn.Linear(
            token_dim,
            num_classes
        )

        self.head_b = nn.Linear(
            token_dim,
            num_classes
        )

    def forward(self, x):
        fa = self.branch_a(x)
        fb = self.branch_b(x)

        a_tokens = self.extractor_a(fa)
        b_tokens = self.extractor_b(fb)

        a_tokens, b_tokens = self.cross_attention(
            a_tokens,
            b_tokens
        )

        fused, gate = self.fusion(
            a_tokens,
            b_tokens
        )

        logits = self.classifier(fused)

        logits_a = self.head_a(
            a_tokens.mean(dim=1)
        )

        logits_b = self.head_b(
            b_tokens.mean(dim=1)
        )

        return {
            "logits": logits,
            "logits_a": logits_a,
            "logits_b": logits_b,
            "features_a": a_tokens.mean(dim=1),
            "features_b": b_tokens.mean(dim=1),
            "fused_features": fused,
            "gate": gate
        }

# ================================================================
# 19. MODEL INITIALIZATION
# ================================================================
model = RLDAFCNN(
    num_classes=NUM_CLASSES,
    token_dim=384,
    pretrained=USE_PRETRAINED
)

if N_GPUS > 1:
    print(f"\nUsing DataParallel on {N_GPUS} GPUs.")
    model = nn.DataParallel(model)

model = model.to(DEVICE)

# ================================================================
# 20. PARAMETER SUMMARY
# ================================================================
def unwrap_model(model):
    return model.module if isinstance(model, nn.DataParallel) else model

def count_parameters(model):
    return sum(
        p.numel() for p in model.parameters()
    )

def count_trainable_parameters(model):
    return sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )

total_params = count_parameters(model)
trainable_params = count_trainable_parameters(model)

print("\nTotal parameters     :", f"{total_params:,}")
print("Trainable parameters :", f"{trainable_params:,}")

# ================================================================
# 21. CLASS WEIGHTS
# ================================================================
train_targets = train_df["label"].values

class_weights_np = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(NUM_CLASSES),
    y=train_targets
)

# Moderate square-root compression prevents the rarest class
# from dominating optimization.
class_weights_np = np.sqrt(class_weights_np)
class_weights_np = (
    class_weights_np /
    class_weights_np.mean()
)

class_weights = torch.tensor(
    class_weights_np,
    dtype=torch.float32,
    device=DEVICE
)

print("\nClass weights:")
for cls, w in zip(CLASS_NAMES, class_weights_np):
    print(f"  {cls:15s}: {w:.4f}")

# ================================================================
# 22. LOSSES
# ================================================================
criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    label_smoothing=LABEL_SMOOTHING
)

def consistency_loss(a, b):
    """
    Cosine distance between branch-level representations.
    It is intentionally small so that the branches can still
    learn complementary information.
    """
    a = F.normalize(a, dim=1)
    b = F.normalize(b, dim=1)

    return (1.0 - (a * b).sum(dim=1)).mean()

def compute_total_loss(outputs, targets):
    main_loss = criterion(
        outputs["logits"],
        targets
    )

    aux_a = criterion(
        outputs["logits_a"],
        targets
    )

    aux_b = criterion(
        outputs["logits_b"],
        targets
    )

    cons = consistency_loss(
        outputs["features_a"],
        outputs["features_b"]
    )

    total = (
        main_loss
        + LAMBDA_AUX * (aux_a + aux_b)
        + LAMBDA_CONSISTENCY * cons
    )

    return total, {
        "main": float(main_loss.detach().cpu()),
        "aux_a": float(aux_a.detach().cpu()),
        "aux_b": float(aux_b.detach().cpu()),
        "consistency": float(cons.detach().cpu()),
    }

# ================================================================
# 23. OPTIMIZER WITH DIFFERENTIAL LR
# ================================================================
base_model = unwrap_model(model)

backbone_params = []
head_params = []

backbone_names = [
    "branch_a",
    "branch_b",
]

for name, p in base_model.named_parameters():
    if not p.requires_grad:
        continue

    if any(name.startswith(x) for x in backbone_names):
        backbone_params.append(p)
    else:
        head_params.append(p)

optimizer = optim.AdamW(
    [
        {
            "params": backbone_params,
            "lr": BACKBONE_LR
        },
        {
            "params": head_params,
            "lr": HEAD_LR
        }
    ],
    weight_decay=WEIGHT_DECAY
)

scheduler = optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=EPOCHS
)

scaler = torch.cuda.amp.GradScaler(
    enabled=torch.cuda.is_available()
)

# ================================================================
# 24. METRIC HELPER
# ================================================================
def classification_metrics(y_true, y_pred):
    return {
        "accuracy": accuracy_score(
            y_true, y_pred
        ),
        "balanced_accuracy": balanced_accuracy_score(
            y_true, y_pred
        ),
        "macro_precision": precision_score(
            y_true, y_pred,
            average="macro",
            zero_division=0
        ),
        "macro_recall": recall_score(
            y_true, y_pred,
            average="macro",
            zero_division=0
        ),
        "macro_f1": f1_score(
            y_true, y_pred,
            average="macro",
            zero_division=0
        ),
        "weighted_f1": f1_score(
            y_true, y_pred,
            average="weighted",
            zero_division=0
        ),
        "mcc": matthews_corrcoef(
            y_true, y_pred
        ),
        "cohen_kappa": cohen_kappa_score(
            y_true, y_pred
        ),
    }

# ================================================================
# 25. TRAIN / VALIDATION EPOCH
# ================================================================
def run_epoch(
    model,
    loader,
    optimizer=None,
    training=True
):
    model.train(training)

    total_loss_value = 0.0
    n_samples = 0

    y_true = []
    y_pred = []

    start_time = time.time()

    if training:
        optimizer.zero_grad(set_to_none=True)

    for step, batch in enumerate(loader):
        images, targets, _ = batch

        images = images.to(
            DEVICE,
            non_blocking=True
        )
        targets = targets.to(
            DEVICE,
            non_blocking=True
        )

        with torch.cuda.amp.autocast(
            enabled=torch.cuda.is_available()
        ):
            outputs = model(images)

            loss, _ = compute_total_loss(
                outputs,
                targets
            )

            if training:
                loss_for_backward = (
                    loss / GRAD_ACCUM_STEPS
                )
            else:
                loss_for_backward = loss

        if training:
            scaler.scale(
                loss_for_backward
            ).backward()

            should_step = (
                ((step + 1) % GRAD_ACCUM_STEPS == 0)
                or ((step + 1) == len(loader))
            )

            if should_step:
                scaler.unscale_(optimizer)

                torch.nn.utils.clip_grad_norm_(
                    model.parameters(),
                    max_norm=1.0
                )

                scaler.step(optimizer)
                scaler.update()

                optimizer.zero_grad(
                    set_to_none=True
                )

        batch_size = images.size(0)

        total_loss_value += (
            loss.detach().float().cpu().item()
            * batch_size
        )

        n_samples += batch_size

        predictions = outputs["logits"].argmax(
            dim=1
        )

        y_true.extend(
            targets.detach().cpu().numpy()
        )

        y_pred.extend(
            predictions.detach().cpu().numpy()
        )

    metrics = classification_metrics(
        np.array(y_true),
        np.array(y_pred)
    )

    metrics["loss"] = (
        total_loss_value / max(n_samples, 1)
    )

    metrics["time_sec"] = time.time() - start_time

    return metrics

# ================================================================
# 26. TRAINING LOOP
# ================================================================
checkpoint_path = (
    OUTPUT_DIR /
    "RLDAF_CNN_best.pth"
)

history = []
best_f1 = -np.inf
best_state = None
wait = 0

print("\n" + "=" * 80)
print("START TRAINING")
print("=" * 80)

for epoch in range(1, EPOCHS + 1):
    start = time.time()

    train_metrics = run_epoch(
        model,
        train_loader,
        optimizer=optimizer,
        training=True
    )

    with torch.no_grad():
        val_metrics = run_epoch(
            model,
            val_loader,
            optimizer=None,
            training=False
        )

    scheduler.step()

    current_lr_backbone = optimizer.param_groups[0]["lr"]
    current_lr_head = optimizer.param_groups[1]["lr"]

    row = {
        "epoch": epoch,
        "lr_backbone": current_lr_backbone,
        "lr_head": current_lr_head,
        "epoch_time_sec": time.time() - start
    }

    for k, v in train_metrics.items():
        row[f"train_{k}"] = v

    for k, v in val_metrics.items():
        row[f"val_{k}"] = v

    history.append(row)

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train Loss {train_metrics['loss']:.4f} | "
        f"Val Loss {val_metrics['loss']:.4f} | "
        f"Train F1 {train_metrics['macro_f1']:.4f} | "
        f"Val F1 {val_metrics['macro_f1']:.4f}"
    )

    if val_metrics["macro_f1"] > best_f1:
        best_f1 = val_metrics["macro_f1"]
        best_state = copy.deepcopy(
            unwrap_model(model).state_dict()
        )

        torch.save(
            {
                "state_dict": best_state,
                "best_val_macro_f1": best_f1,
                "class_names": CLASS_NAMES,
                "image_size": IMAGE_SIZE,
                "seed": SEED
            },
            checkpoint_path
        )

        wait = 0
        print("  -> Best checkpoint saved.")
    else:
        wait += 1

    if wait >= PATIENCE:
        print("Early stopping.")
        break

if best_state is not None:
    unwrap_model(model).load_state_dict(
        best_state
    )

history_df = pd.DataFrame(history)
history_df.to_csv(
    OUTPUT_DIR / "training_history.csv",
    index=False
)

# ================================================================
# 27. TRAINING CURVES
# ================================================================
def plot_history(
    history_df,
    metric,
    ylabel,
    filename
):
    plt.figure(figsize=(8, 5))

    plt.plot(
        history_df["epoch"],
        history_df[f"train_{metric}"],
        label="Train"
    )

    plt.plot(
        history_df["epoch"],
        history_df[f"val_{metric}"],
        label="Validation"
    )

    plt.xlabel("Epoch")
    plt.ylabel(ylabel)
    plt.title(f"Training and Validation {ylabel}")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / filename,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

plot_history(
    history_df,
    "loss",
    "Loss",
    "curve_loss.png"
)

plot_history(
    history_df,
    "accuracy",
    "Accuracy",
    "curve_accuracy.png"
)

plot_history(
    history_df,
    "macro_f1",
    "Macro-F1",
    "curve_macro_f1.png"
)

# LR curve
plt.figure(figsize=(8, 5))
plt.plot(
    history_df["epoch"],
    history_df["lr_backbone"],
    label="Backbone LR"
)
plt.plot(
    history_df["epoch"],
    history_df["lr_head"],
    label="Head LR"
)
plt.xlabel("Epoch")
plt.ylabel("Learning rate")
plt.title("Learning Rate Schedule")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIR / "curve_learning_rate.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ================================================================
# 28. PREDICTION FUNCTION
# ================================================================
@torch.no_grad()
def predict_model(model, loader):
    model.eval()

    all_true = []
    all_pred = []
    all_prob = []
    all_paths = []
    all_features = []
    all_a = []
    all_b = []
    all_gates = []

    for images, targets, paths in loader:
        images = images.to(
            DEVICE,
            non_blocking=True
        )

        outputs = model(images)

        probs = torch.softmax(
            outputs["logits"],
            dim=1
        )

        preds = probs.argmax(dim=1)

        all_true.extend(
            targets.numpy()
        )
        all_pred.extend(
            preds.cpu().numpy()
        )
        all_prob.append(
            probs.cpu().numpy()
        )
        all_paths.extend(paths)

        all_features.append(
            outputs["fused_features"].cpu().numpy()
        )

        all_a.append(
            outputs["features_a"].cpu().numpy()
        )

        all_b.append(
            outputs["features_b"].cpu().numpy()
        )

        all_gates.append(
            outputs["gate"].cpu().numpy()
        )

    return {
        "true": np.array(all_true),
        "pred": np.array(all_pred),
        "prob": np.concatenate(all_prob, axis=0),
        "paths": all_paths,
        "features": np.concatenate(all_features, axis=0),
        "features_a": np.concatenate(all_a, axis=0),
        "features_b": np.concatenate(all_b, axis=0),
        "gates": np.concatenate(all_gates, axis=0),
    }

# ================================================================
# 29. TEST SET EVALUATION
# ================================================================
print("\n" + "=" * 80)
print("FINAL TEST EVALUATION")
print("=" * 80)

test_output = predict_model(
    model,
    test_loader
)

y_true = test_output["true"]
y_pred = test_output["pred"]
y_prob = test_output["prob"]

main_metrics = classification_metrics(
    y_true,
    y_pred
)

# ROC-AUC and PR-AUC
y_true_bin = label_binarize(
    y_true,
    classes=np.arange(NUM_CLASSES)
)

try:
    roc_auc_macro = roc_auc_score(
        y_true_bin,
        y_prob,
        average="macro",
        multi_class="ovr"
    )
except Exception:
    roc_auc_macro = np.nan

try:
    pr_auc_macro = average_precision_score(
        y_true_bin,
        y_prob,
        average="macro"
    )
except Exception:
    pr_auc_macro = np.nan

main_metrics["roc_auc_macro_ovr"] = roc_auc_macro
main_metrics["pr_auc_macro"] = pr_auc_macro

metrics_df = pd.DataFrame([
    {
        "metric": k,
        "value": v
    }
    for k, v in main_metrics.items()
])

metrics_df.to_csv(
    OUTPUT_DIR / "test_metrics.csv",
    index=False
)

print(metrics_df.to_string(index=False))

# ================================================================
# 30. CLASSIFICATION REPORT
# ================================================================
report_dict = classification_report(
    y_true,
    y_pred,
    target_names=CLASS_NAMES,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report_dict).T
report_df.to_csv(
    OUTPUT_DIR / "classification_report.csv"
)

print("\nClassification report:")
print(
    classification_report(
        y_true,
        y_pred,
        target_names=CLASS_NAMES,
        zero_division=0
    )
)

# ================================================================
# 31. PER-CLASS TP TN FP FN + SENSITIVITY ETC.
# ================================================================
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES)
)

per_class_rows = []

total = cm.sum()

for i, cls in enumerate(CLASS_NAMES):
    tp = cm[i, i]
    fn = cm[i, :].sum() - tp
    fp = cm[:, i].sum() - tp
    tn = total - tp - fn - fp

    sensitivity = (
        tp / (tp + fn)
        if (tp + fn) else 0.0
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) else 0.0
    )

    precision_value = (
        tp / (tp + fp)
        if (tp + fp) else 0.0
    )

    npv = (
        tn / (tn + fn)
        if (tn + fn) else 0.0
    )

    f1 = (
        2 * precision_value * sensitivity /
        (precision_value + sensitivity)
        if (precision_value + sensitivity)
        else 0.0
    )

    per_class_rows.append({
        "class": cls,
        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "sensitivity_recall": sensitivity,
        "specificity": specificity,
        "precision": precision_value,
        "NPV": npv,
        "F1": f1
    })

per_class_df = pd.DataFrame(per_class_rows)

per_class_df.to_csv(
    OUTPUT_DIR / "per_class_detailed_metrics.csv",
    index=False
)

print("\nPer-class detailed metrics:")
print(per_class_df.to_string(index=False))

# ================================================================
# 32. CONFUSION MATRICES
# ================================================================
plt.figure(figsize=(9, 7))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES,
    cmap="Blues"
)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.tight_layout()
plt.savefig(
    OUTPUT_DIR / "confusion_matrix_raw.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

cm_norm = (
    cm.astype(np.float64) /
    np.maximum(
        cm.sum(axis=1, keepdims=True),
        1
    )
)

plt.figure(figsize=(9, 7))
sns.heatmap(
    cm_norm,
    annot=True,
    fmt=".2f",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES,
    cmap="Blues",
    vmin=0,
    vmax=1
)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Normalized Confusion Matrix")
plt.tight_layout()
plt.savefig(
    OUTPUT_DIR / "confusion_matrix_normalized.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ================================================================
# 33. ROC CURVES
# ================================================================
plt.figure(figsize=(9, 7))

for i, cls in enumerate(CLASS_NAMES):
    fpr, tpr, _ = roc_curve(
        y_true_bin[:, i],
        y_prob[:, i]
    )

    auc_i = roc_auc_score(
        y_true_bin[:, i],
        y_prob[:, i]
    )

    plt.plot(
        fpr,
        tpr,
        label=f"{cls} (AUC={auc_i:.3f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("One-vs-Rest ROC Curves")
plt.legend(fontsize=8)
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIR / "roc_curves.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ================================================================
# 34. PRECISION-RECALL CURVES
# ================================================================
plt.figure(figsize=(9, 7))

for i, cls in enumerate(CLASS_NAMES):
    precision_i, recall_i, _ = precision_recall_curve(
        y_true_bin[:, i],
        y_prob[:, i]
    )

    ap_i = average_precision_score(
        y_true_bin[:, i],
        y_prob[:, i]
    )

    plt.plot(
        recall_i,
        precision_i,
        label=f"{cls} (AP={ap_i:.3f})"
    )

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("One-vs-Rest Precision-Recall Curves")
plt.legend(fontsize=8)
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIR / "pr_curves.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ================================================================
# 35. CONFIDENCE ANALYSIS
# ================================================================
confidence = y_prob.max(axis=1)
correct = y_true == y_pred

confidence_df = pd.DataFrame({
    "path": test_output["paths"],
    "true_class": [
        CLASS_NAMES[x] for x in y_true
    ],
    "predicted_class": [
        CLASS_NAMES[x] for x in y_pred
    ],
    "confidence": confidence,
    "correct": correct
})

confidence_df.to_csv(
    OUTPUT_DIR / "confidence_analysis.csv",
    index=False
)

plt.figure(figsize=(8, 5))
plt.hist(
    confidence[correct],
    bins=20,
    alpha=0.65,
    label="Correct"
)
plt.hist(
    confidence[~correct],
    bins=20,
    alpha=0.65,
    label="Incorrect"
)
plt.xlabel("Maximum predicted probability")
plt.ylabel("Number of samples")
plt.title("Prediction Confidence Distribution")
plt.legend()
plt.tight_layout()
plt.savefig(
    OUTPUT_DIR / "confidence_distribution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ================================================================
# 36. MISCLASSIFICATION PAIRS
# ================================================================
error_df = confidence_df[
    confidence_df["correct"] == False
].copy()

pair_counts = (
    error_df
    .groupby(
        ["true_class", "predicted_class"]
    )
    .size()
    .reset_index(name="count")
    .sort_values("count", ascending=False)
)

pair_counts.to_csv(
    OUTPUT_DIR / "misclassification_pairs.csv",
    index=False
)

print("\nMost frequent misclassification pairs:")
print(pair_counts.head(15).to_string(index=False))

# ================================================================
# 37. ADAPTIVE GATE ANALYSIS
# ================================================================
gate_values = test_output["gates"]

gate_df = pd.DataFrame(
    gate_values,
    columns=[
        "ConvNeXtV2_gate",
        "EfficientNetV2_gate"
    ]
)

gate_df["true_class"] = [
    CLASS_NAMES[x] for x in y_true
]

gate_df["predicted_class"] = [
    CLASS_NAMES[x] for x in y_pred
]

gate_df["correct"] = correct

gate_df.to_csv(
    OUTPUT_DIR / "adaptive_gate_values.csv",
    index=False
)

gate_summary = (
    gate_df
    .groupby("true_class")
    [
        [
            "ConvNeXtV2_gate",
            "EfficientNetV2_gate"
        ]
    ]
    .agg(["mean", "std", "median"])
)

gate_summary.to_csv(
    OUTPUT_DIR / "gate_by_class.csv"
)

# Simple mean gate figure
gate_means = (
    gate_df
    .groupby("true_class")[
        [
            "ConvNeXtV2_gate",
            "EfficientNetV2_gate"
        ]
    ]
    .mean()
    .reindex(CLASS_NAMES)
)

gate_means.plot(
    kind="bar",
    figsize=(10, 5)
)
plt.ylabel("Mean gate value")
plt.xlabel("Class")
plt.title("Adaptive Dual-CNN Branch Preference")
plt.xticks(rotation=30, ha="right")
plt.legend()
plt.tight_layout()
plt.savefig(
    OUTPUT_DIR / "gate_by_class.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ================================================================
# 38. FEATURE SPACE: PCA
# ================================================================
feature_df = test_df.copy()

if len(feature_df) > MAX_FEATURE_IMAGES:
    keep_idx = np.random.RandomState(SEED).choice(
        len(feature_df),
        MAX_FEATURE_IMAGES,
        replace=False
    )
    feature_df = feature_df.iloc[keep_idx].reset_index(drop=True)

feature_ds = RiceLeafDataset(
    feature_df,
    eval_transform
)

feature_loader = DataLoader(
    feature_ds,
    batch_size=VAL_BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

feature_output = predict_model(
    model,
    feature_loader
)

features = feature_output["features"]
feature_labels = feature_output["true"]

if RUN_PCA:
    pca2 = PCA(n_components=2)
    pca_features = pca2.fit_transform(features)

    plt.figure(figsize=(8, 7))

    for i, cls in enumerate(CLASS_NAMES):
        mask = feature_labels == i

        plt.scatter(
            pca_features[mask, 0],
            pca_features[mask, 1],
            s=16,
            alpha=0.65,
            label=cls
        )

    plt.xlabel("PC1")
    plt.ylabel("PC2")
    plt.title("PCA of Fused Feature Space")
    plt.legend(fontsize=8)
    plt.grid(alpha=0.20)
    plt.tight_layout()
    plt.savefig(
        OUTPUT_DIR / "pca_feature_space.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

# ================================================================
# 39. FEATURE SPACE: t-SNE
# ================================================================
if RUN_TSNE and len(features) >= 20:
    perplexity = min(
        30,
        max(5, (len(features) - 1) // 3)
    )

    pca_dim = min(
        50,
        features.shape[1],
        len(features) - 1
    )

    reduced = PCA(
        n_components=pca_dim
    ).fit_transform(features)

    tsne = TSNE(
        n_components=2,
        random_state=SEED,
        perplexity=perplexity,
        init="pca",
        learning_rate="auto"
    )

    tsne_features = tsne.fit_transform(
        reduced
    )

    plt.figure(figsize=(8, 7))

    for i, cls in enumerate(CLASS_NAMES):
        mask = feature_labels == i

        plt.scatter(
            tsne_features[mask, 0],
            tsne_features[mask, 1],
            s=16,
            alpha=0.65,
            label=cls
        )

    plt.xlabel("t-SNE 1")
    plt.ylabel("t-SNE 2")
    plt.title("t-SNE of Fused Feature Space")
    plt.legend(fontsize=8)
    plt.grid(alpha=0.20)
    plt.tight_layout()
    plt.savefig(
        OUTPUT_DIR / "tsne_feature_space.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

# ================================================================
# 40. FEATURE SPACE: UMAP
# ================================================================
if RUN_UMAP and len(features) >= 20:
    try:
        import umap.umap_ as umap

        reducer = umap.UMAP(
            n_components=2,
            random_state=SEED,
            n_neighbors=min(
                15,
                len(features) - 1
            ),
            min_dist=0.15
        )

        umap_features = reducer.fit_transform(
            features
        )

        plt.figure(figsize=(8, 7))

        for i, cls in enumerate(CLASS_NAMES):
            mask = feature_labels == i

            plt.scatter(
                umap_features[mask, 0],
                umap_features[mask, 1],
                s=16,
                alpha=0.65,
                label=cls
            )

        plt.xlabel("UMAP 1")
        plt.ylabel("UMAP 2")
        plt.title("UMAP of Fused Feature Space")
        plt.legend(fontsize=8)
        plt.grid(alpha=0.20)
        plt.tight_layout()
        plt.savefig(
            OUTPUT_DIR / "umap_feature_space.png",
            dpi=300,
            bbox_inches="tight"
        )
        plt.show()
        plt.close()

    except Exception as e:
        print("UMAP skipped:", e)

# ================================================================
# 41. CALIBRATION / ECE
# ================================================================
def expected_calibration_error(
    y_true,
    y_prob,
    n_bins=15
):
    confidence = np.max(
        y_prob,
        axis=1
    )

    predictions = np.argmax(
        y_prob,
        axis=1
    )

    correctness = (
        predictions == y_true
    ).astype(float)

    bins = np.linspace(
        0.0,
        1.0,
        n_bins + 1
    )

    ece = 0.0
    rows = []

    for i in range(n_bins):
        lo = bins[i]
        hi = bins[i + 1]

        if i == n_bins - 1:
            mask = (
                (confidence >= lo) &
                (confidence <= hi)
            )
        else:
            mask = (
                (confidence >= lo) &
                (confidence < hi)
            )

        if mask.sum() == 0:
            continue

        bin_acc = correctness[mask].mean()
        bin_conf = confidence[mask].mean()
        fraction = mask.mean()

        ece += (
            fraction *
            abs(bin_acc - bin_conf)
        )

        rows.append({
            "bin_lower": lo,
            "bin_upper": hi,
            "accuracy": bin_acc,
            "confidence": bin_conf,
            "fraction": fraction
        })

    return ece, pd.DataFrame(rows)

ece, calibration_df = expected_calibration_error(
    y_true,
    y_prob,
    n_bins=15
)

calibration_df.to_csv(
    OUTPUT_DIR / "calibration_bins.csv",
    index=False
)

# Multiclass Brier score:
# mean squared error over one-hot target vectors.
brier = np.mean(
    np.sum(
        (y_prob - y_true_bin) ** 2,
        axis=1
    )
)

calibration_metrics = pd.DataFrame([
    {"metric": "ECE", "value": ece},
    {"metric": "Brier_score_multiclass", "value": brier}
])

calibration_metrics.to_csv(
    OUTPUT_DIR / "calibration_metrics.csv",
    index=False
)

print("\nCalibration:")
print(calibration_metrics.to_string(index=False))

# Reliability diagram
plt.figure(figsize=(7, 6))

if len(calibration_df):
    plt.plot(
        calibration_df["confidence"],
        calibration_df["accuracy"],
        marker="o",
        label="RLDAF-CNN"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Perfect calibration"
)

plt.xlabel("Mean predicted confidence")
plt.ylabel("Empirical accuracy")
plt.title(f"Reliability Diagram (ECE={ece:.4f})")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIR / "reliability_diagram.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ================================================================
# 42. BOOTSTRAP CONFIDENCE INTERVALS
# ================================================================
def bootstrap_metric_ci(
    y_true,
    y_pred,
    metric_fn,
    rounds=1000,
    seed=42
):
    rng = np.random.default_rng(seed)
    n = len(y_true)

    values = []

    for _ in range(rounds):
        idx = rng.integers(
            0,
            n,
            size=n
        )

        try:
            value = metric_fn(
                y_true[idx],
                y_pred[idx]
            )

            if np.isfinite(value):
                values.append(value)
        except Exception:
            pass

    if not values:
        return np.nan, np.nan, np.nan

    values = np.array(values)

    return (
        float(np.mean(values)),
        float(np.percentile(values, 2.5)),
        float(np.percentile(values, 97.5))
    )

bootstrap_rows = []

bootstrap_metrics = {
    "accuracy": accuracy_score,
    "balanced_accuracy": balanced_accuracy_score,
    "macro_precision": lambda a, b: precision_score(
        a, b, average="macro", zero_division=0
    ),
    "macro_recall": lambda a, b: recall_score(
        a, b, average="macro", zero_division=0
    ),
    "macro_f1": lambda a, b: f1_score(
        a, b, average="macro", zero_division=0
    ),
    "mcc": matthews_corrcoef,
    "cohen_kappa": cohen_kappa_score,
}

for name, fn in bootstrap_metrics.items():
    mean_v, low_v, high_v = bootstrap_metric_ci(
        y_true,
        y_pred,
        fn,
        rounds=BOOTSTRAP_ROUNDS,
        seed=SEED
    )

    bootstrap_rows.append({
        "metric": name,
        "mean": mean_v,
        "CI_2.5%": low_v,
        "CI_97.5%": high_v
    })

bootstrap_df = pd.DataFrame(
    bootstrap_rows
)

bootstrap_df.to_csv(
    OUTPUT_DIR / "bootstrap_95CI.csv",
    index=False
)

print("\nBootstrap 95% CIs:")
print(bootstrap_df.to_string(index=False))

# ================================================================
# 43. COMPUTE MODEL FLOPs / PARAMS
# ================================================================
try:
    from thop import profile

    dummy = torch.randn(
        1, 3, IMAGE_SIZE, IMAGE_SIZE
    ).to(DEVICE)

    base_for_profile = unwrap_model(model)

    was_training = base_for_profile.training
    base_for_profile.eval()

    with torch.no_grad():
        flops, params = profile(
            base_for_profile,
            inputs=(dummy,),
            verbose=False
        )

    if was_training:
        base_for_profile.train()

    print(
        "\nApproximate parameters:",
        f"{params / 1e6:.2f} M"
    )
    print(
        "Approximate FLOPs:",
        f"{flops / 1e9:.2f} G"
    )

    pd.DataFrame([{
        "parameters_M": params / 1e6,
        "FLOPs_G": flops / 1e9,
        "image_size": IMAGE_SIZE
    }]).to_csv(
        OUTPUT_DIR / "complexity.csv",
        index=False
    )

    del dummy

except Exception as e:
    print("THOP complexity calculation skipped:", e)

# ================================================================
# 44. INFERENCE LATENCY / FPS
# ================================================================
def benchmark_latency(
    model,
    image_size,
    repeats=50,
    warmup=10
):
    model.eval()

    x = torch.randn(
        1, 3,
        image_size,
        image_size,
        device=DEVICE
    )

    with torch.no_grad():
        for _ in range(warmup):
            _ = model(x)

        if torch.cuda.is_available():
            torch.cuda.synchronize()

        start = time.perf_counter()

        for _ in range(repeats):
            _ = model(x)

        if torch.cuda.is_available():
            torch.cuda.synchronize()

        elapsed = time.perf_counter() - start

    latency_ms = (
        elapsed / repeats
    ) * 1000.0

    fps = (
        1000.0 / latency_ms
        if latency_ms > 0
        else 0
    )

    return latency_ms, fps

try:
    latency_ms, fps = benchmark_latency(
        model,
        IMAGE_SIZE
    )

    print(
        f"\nInference latency: {latency_ms:.2f} ms/image"
    )
    print(
        f"Inference throughput: {fps:.2f} images/sec"
    )

    pd.DataFrame([{
        "latency_ms_per_image": latency_ms,
        "FPS": fps,
        "image_size": IMAGE_SIZE
    }]).to_csv(
        OUTPUT_DIR / "inference_speed.csv",
        index=False
    )

except Exception as e:
    print("Latency benchmark skipped:", e)

# ================================================================
# 45. GPU MEMORY
# ================================================================
if torch.cuda.is_available():
    for gpu_id in range(torch.cuda.device_count()):
        allocated = (
            torch.cuda.memory_allocated(gpu_id)
            / (1024 ** 3)
        )

        reserved = (
            torch.cuda.memory_reserved(gpu_id)
            / (1024 ** 3)
        )

        print(
            f"GPU {gpu_id}: "
            f"allocated={allocated:.2f} GB, "
            f"reserved={reserved:.2f} GB"
        )

# ================================================================
# 46. OPTIONAL ROBUSTNESS ANALYSIS
# ================================================================
# This evaluates the already-trained model under controlled
# image corruptions. It does NOT retrain the model.
if RUN_ROBUSTNESS:

    print("\nRunning robustness analysis...")

    robustness_transforms = {
        "normal": eval_transform,

        "brightness_low": transforms.Compose([
            transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
            transforms.ColorJitter(brightness=(0.55, 0.55)),
            transforms.ToTensor(),
            transforms.Normalize(
                IMAGENET_MEAN,
                IMAGENET_STD
            )
        ]),

        "contrast_low": transforms.Compose([
            transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
            transforms.ColorJitter(contrast=(0.55, 0.55)),
            transforms.ToTensor(),
            transforms.Normalize(
                IMAGENET_MEAN,
                IMAGENET_STD
            )
        ]),
    }

    robustness_rows = []

    for name, tfm in robustness_transforms.items():

        ds = RiceLeafDataset(
            test_df,
            tfm
        )

        loader = DataLoader(
            ds,
            batch_size=VAL_BATCH_SIZE,
            shuffle=False,
            num_workers=NUM_WORKERS
        )

        out = predict_model(
            model,
            loader
        )

        m = classification_metrics(
            out["true"],
            out["pred"]
        )

        robustness_rows.append({
            "condition": name,
            **m
        })

    robustness_df = pd.DataFrame(
        robustness_rows
    )

    robustness_df.to_csv(
        OUTPUT_DIR / "robustness_results.csv",
        index=False
    )

    print(robustness_df.to_string(index=False))

# ================================================================
# 47. OPTIONAL BASELINES
# ================================================================
# These are intentionally disabled by default.
#
# For a publication-grade comparison, use exactly the SAME:
#   * cleaned image set
#   * 70/15/15 split
#   * image resolution
#   * training seed
#   * evaluation protocol
#
# Suggested baseline models:
#   1. ConvNeXtV2-Tiny
#   2. EfficientNetV2-S
#   3. simple feature concatenation
#   4. proposed RLDAF-CNN
#
# Implement these after the main model has completed successfully.

if RUN_BASELINES:
    print(
        "\nBaseline section is enabled. "
        "Add identical-training baseline configurations here."
    )

# ================================================================
# 48. OPTIONAL ABLATION STUDY
# ================================================================
# Recommended ablations for the paper:
#
# A0: ConvNeXtV2 only
# A1: EfficientNetV2 only
# A2: Dual CNN + concatenation
# A3: Dual CNN + CBAM
# A4: Dual CNN + CBAM + cross-attention
# A5: Full RLDAF-CNN + adaptive gated fusion
#
# Keep the data split and training budget identical.

if RUN_ABLATION:
    print(
        "\nAblation section is enabled. "
        "Run A0-A5 using the same split and seed."
    )

# ================================================================
# 49. SAVE EXPERIMENT CONFIGURATION
# ================================================================
config = {
    "seed": SEED,
    "dataset_root": str(DATASET_ROOT),
    "image_root": str(IMAGE_ROOT),
    "classes": CLASS_NAMES,
    "num_classes": NUM_CLASSES,
    "image_size": IMAGE_SIZE,
    "batch_size": BATCH_SIZE,
    "val_batch_size": VAL_BATCH_SIZE,
    "gradient_accumulation": GRAD_ACCUM_STEPS,
    "epochs": EPOCHS,
    "patience": PATIENCE,
    "backbone_lr": BACKBONE_LR,
    "head_lr": HEAD_LR,
    "weight_decay": WEIGHT_DECAY,
    "label_smoothing": LABEL_SMOOTHING,
    "lambda_aux": LAMBDA_AUX,
    "lambda_consistency": LAMBDA_CONSISTENCY,
    "pretrained": USE_PRETRAINED,
    "train_size": len(train_df),
    "validation_size": len(val_df),
    "test_size": len(test_df),
    "total_clean_images": len(df),
    "exact_duplicates_removed": before_dup - len(df),
    "total_parameters": total_params,
    "trainable_parameters": trainable_params,
}

with open(
    OUTPUT_DIR / "experiment_config.json",
    "w"
) as f:
    json.dump(
        config,
        f,
        indent=4
    )

# ================================================================
# 50. FINAL SUMMARY
# ================================================================
print("\n" + "=" * 80)
print("EXPERIMENT FINISHED")
print("=" * 80)

print("Model: RLDAF-CNN")
print("Branch A: ConvNeXtV2-Tiny")
print("Branch B: EfficientNetV2-S")
print("Attention: CBAM + bidirectional cross-branch attention")
print("Fusion: sample-adaptive gated fusion")
print("Images used: Original images ONLY")
print("Annotations/masks/ROI: NOT USED")
print(f"Clean images: {len(df)}")
print(f"Train: {len(train_df)}")
print(f"Validation: {len(val_df)}")
print(f"Test: {len(test_df)}")
print(f"Best validation Macro-F1: {best_f1:.4f}")

print("\nFinal test metrics:")
for k, v in main_metrics.items():
    print(f"{k:25s}: {v:.6f}")

print("\nAll outputs saved to:")
print(OUTPUT_DIR)

# ================================================================
# 51. OPTIONAL ZIP OF ALL RESULTS
# ================================================================
import shutil

zip_path = shutil.make_archive(
    str(OUTPUT_DIR),
    "zip",
    root_dir=OUTPUT_DIR
)

print("\nResults ZIP:")
print(zip_path)

# ================================================================
# END
# ================================================================
```
